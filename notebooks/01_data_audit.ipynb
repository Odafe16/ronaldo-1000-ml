# DATA AUDIT

```python
import pandas as pd


DATA_PATH = "../data/raw/ronaldo_complete_ml_dataset.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nTotal columns:", len(df.columns))
print("\nColumns:")
print(df.columns.tolist())
```

## Core reconciliation and chronology checks

```python


# Convert the date column from text/string into a proper
# pandas datetime type so chronological checks work correctly.
df["date"] = pd.to_datetime(df["date"])

# Sort the dataset by match date from oldest to newest.
# reset_index(drop=True) gives the dataframe a clean 0...N index.
df = df.sort_values("date").reset_index(drop=True)


# Number of matches/appearances in the dataset.
print("Total appearances:", len(df))

# Total career goals represented by the target column.
print("Total goals:", df["target_goals"].sum())

# Check how many unique match IDs exist.
print("Unique match IDs:", df["match_id"].nunique())

# Check whether any match_id occurs more than once.
print("Duplicate match IDs:", df["match_id"].duplicated().sum())



# Earliest match in the dataset.
print("First match date:", df["date"].min())

# Latest match in the dataset.
print("Last match date:", df["date"].max())

# True means every row is ordered chronologically.
print("Dates monotonically increasing:", df["date"].is_monotonic_increasing)



# HARD ASSERTIONS
# These should all pass silently.
# If one fails, Python will raise an AssertionError.

assert len(df) == 1333, "Unexpected number of appearances."

assert df["target_goals"].sum() == 978, \
    "Career goal total does not reconcile to 978."

assert df["match_id"].is_unique, \
    "Duplicate match IDs exist."

assert df["date"].is_monotonic_increasing, \
    "Dataset is not ordered chronologically."


print("\nAll Step 1B core checks passed.")
```

## Target integrity

```python
print("Goal-count distribution:")
print(df["target_goals"].value_counts().sort_index())

print("\nTarget value ranges:")
print("target_goals:", sorted(df["target_goals"].unique()))
print("target_scored:", sorted(df["target_scored"].unique()))
print("target_multi_goal:", sorted(df["target_multi_goal"].unique()))

# Check that the derived binary targets agree with target_goals
scored_mismatch = df[
    df["target_scored"] != (df["target_goals"] > 0).astype(int)
]

multi_mismatch = df[
    df["target_multi_goal"] != (df["target_goals"] >= 2).astype(int)
]

print("\nNegative goal rows:", (df["target_goals"] < 0).sum())
print("target_scored mismatches:", len(scored_mismatch))
print("target_multi_goal mismatches:", len(multi_mismatch))
```

## Missing-value audit

```python
# Missing-value audit

missing = (
    df.isna()
      .sum()
      .to_frame("missing_count")
)

missing["missing_pct"] = (missing["missing_count"] / len(df) * 100).round(2)

# Show only columns containing missing values
missing = missing[missing["missing_count"] > 0].sort_values(
    "missing_count",
    ascending=False
)

print("Columns with missing values:", len(missing))
print("Columns with no missing values:", df.shape[1] - len(missing))

display(missing)
```

```python
# Check whether H2H missingness is structurally expected

h2h_check = df[
    [
        "date",
        "opponent",
        "opponent_prior_meetings_pre",
        "opponent_score_data_h2h_matches_pre",
        "ronaldo_goals_per_game_vs_opponent_pre",
        "opponent_goals_for_per_h2h_scored_match_pre",
        "opponent_goals_against_per_h2h_scored_match_pre",
        "opponent_clean_sheet_rate_h2h_scored_match_pre",
    ]
].copy()

# Compare missingness with the amount of prior H2H information available
print("Rows with NO prior opponent meetings:")
print((h2h_check["opponent_prior_meetings_pre"] == 0).sum())

print("\nRows with NO prior verified-score H2H matches:")
print((h2h_check["opponent_score_data_h2h_matches_pre"] == 0).sum())

print("\nMissing Ronaldo H2H GPG:")
print(h2h_check["ronaldo_goals_per_game_vs_opponent_pre"].isna().sum())

print("\nMissing score-based opponent H2H stats:")
print(
    h2h_check[
        "opponent_goals_for_per_h2h_scored_match_pre"
    ].isna().sum()
)

# Look for suspicious cases: prior history exists, but the feature is still missing
suspicious_h2h = h2h_check[
    (h2h_check["opponent_score_data_h2h_matches_pre"] > 0)
    & (
        h2h_check[
            "opponent_goals_for_per_h2h_scored_match_pre"
        ].isna()
    )
]

print("\nPrior verified H2H exists but score-based feature is missing:")
print(len(suspicious_h2h))
```

```python
# Find other likely duplicate opponent names

from difflib import SequenceMatcher

opponents = sorted(df["opponent"].dropna().unique())
similar_pairs = []

for i, name_a in enumerate(opponents):
    for name_b in opponents[i + 1:]:
        score = SequenceMatcher(None, name_a.lower(), name_b.lower()).ratio()

        if score >= 0.72:
            similar_pairs.append((name_a, name_b, round(score, 3)))

similar_pairs = (
    pd.DataFrame(
        similar_pairs,
        columns=["opponent_a", "opponent_b", "similarity"]
    )
    .sort_values("similarity", ascending=False)
    .reset_index(drop=True)
)

display(similar_pairs)
```

```python
# Build corrected identity fields without changing raw columns

df["team_corrected"] = df["team"]
df["opponent_corrected"] = df["opponent"]
df["competition_type_corrected"] = df["competition_type"]

# Correct the 16 known team assignments
for idx, corrected_team in team_corrections.items():
    df.loc[idx, "team_corrected"] = corrected_team

# Normalize confirmed opponent aliases
df["opponent_corrected"] = df["opponent"].replace(opponent_alias_map)

# Fix club competition classifications
df.loc[
    df["competition"].isin([
        "UEFA Europa League",
        "FIFA Club World Cup",
        "UEFA Super Cup",
    ]),
    "competition_type_corrected"
] = "continental_club"

# Rebuild match IDs using corrected identities
df["match_id_corrected"] = (
    df["date"].dt.strftime("%Y-%m-%d")
    + "_"
    + df["team_corrected"].str.replace(" ", "_", regex=False)
    + "_"
    + df["opponent_corrected"].str.replace(" ", "_", regex=False)
    + "_"
    + (df.index + 1).astype(str)
)

display(
    df.loc[list(team_corrections.keys()), [
        "match_id",
        "match_id_corrected",
        "team",
        "team_corrected",
        "competition",
        "competition_type",
        "competition_type_corrected",
    ]]
)
```
