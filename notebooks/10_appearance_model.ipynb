```python
# Reconcile Sporting fixtures with canonical Ronaldo appearances


# Work from fresh copies
fixtures = sporting_fixtures.copy()
apps = sporting_apps.copy()

fixtures["date"] = pd.to_datetime(fixtures["date"])
apps["date"] = pd.to_datetime(apps["date"])

# Normalized keys for safe matching
def normalize_text(value):
    return (
        str(value)
        .strip()
        .lower()
        .replace("á", "a")
        .replace("é", "e")
        .replace("í", "i")
        .replace("ó", "o")
        .replace("ú", "u")
        .replace("ã", "a")
        .replace("ç", "c")
    )

fixtures["opponent_key"] = fixtures["opponent"].map(normalize_text)
fixtures["competition_key"] = fixtures["competition"].map(normalize_text)

apps["opponent_key"] = apps["opponent"].map(normalize_text)
apps["competition_key"] = apps["competition"].map(normalize_text)

# Result columns
fixtures["appeared"] = 0
fixtures["canonical_match_id"] = pd.NA
fixtures["canonical_date"] = pd.NaT
fixtures["match_method"] = "unmatched"

used_app_indices = set()

# ------------------------------------------------------------------
# Pass 1 — exact date + opponent + competition
# ------------------------------------------------------------------
for fixture_idx, fixture in fixtures.iterrows():

    candidates = apps[
        (apps["date"] == fixture["date"]) &
        (apps["opponent_key"] == fixture["opponent_key"]) &
        (apps["competition_key"] == fixture["competition_key"])
    ]

    candidates = candidates[
        ~candidates.index.isin(used_app_indices)
    ]

    if len(candidates) == 1:
        app_idx = candidates.index[0]

        fixtures.loc[fixture_idx, "appeared"] = 1
        fixtures.loc[fixture_idx, "canonical_match_id"] = apps.loc[
            app_idx, "match_id"
        ]
        fixtures.loc[fixture_idx, "canonical_date"] = apps.loc[
            app_idx, "date"
        ]
        fixtures.loc[fixture_idx, "match_method"] = "exact"

        used_app_indices.add(app_idx)


# ------------------------------------------------------------------
# Pass 2 — known source-date discrepancies
# Match remaining rows by opponent + competition where date differs
# by no more than 1 day.
# ------------------------------------------------------------------
for fixture_idx, fixture in fixtures[
    fixtures["match_method"] == "unmatched"
].iterrows():

    candidates = apps[
        (apps["opponent_key"] == fixture["opponent_key"]) &
        (apps["competition_key"] == fixture["competition_key"]) &
        (~apps.index.isin(used_app_indices))
    ].copy()

    if candidates.empty:
        continue

    candidates["date_difference_days"] = (
        candidates["date"] - fixture["date"]
    ).abs().dt.days

    candidates = candidates[
        candidates["date_difference_days"] <= 1
    ]

    if len(candidates) == 1:
        app_idx = candidates.index[0]

        fixtures.loc[fixture_idx, "appeared"] = 1
        fixtures.loc[fixture_idx, "canonical_match_id"] = apps.loc[
            app_idx, "match_id"
        ]
        fixtures.loc[fixture_idx, "canonical_date"] = apps.loc[
            app_idx, "date"
        ]
        fixtures.loc[fixture_idx, "match_method"] = "date_discrepancy"

        used_app_indices.add(app_idx)


# ------------------------------------------------------------------
# Audit
# ------------------------------------------------------------------
matched = fixtures["appeared"].sum()

unmatched_apps = apps[
    ~apps.index.isin(used_app_indices)
].copy()

non_appearances = fixtures[
    fixtures["appeared"] == 0
][
    ["date", "opponent", "competition", "venue"]
].copy()

date_discrepancies = fixtures[
    fixtures["match_method"] == "date_discrepancy"
][
    [
        "date",
        "canonical_date",
        "opponent",
        "competition",
        "canonical_match_id",
    ]
].copy()

print("SPORTING APPEARANCE RECONCILIATION")
print("=" * 90)
print(f"Complete Sporting fixtures: {len(fixtures)}")
print(f"Canonical Ronaldo appearances: {len(apps)}")
print(f"Successfully reconciled: {matched}")
print(f"Unmatched canonical appearances: {len(unmatched_apps)}")
print(f"Candidate non-appearances: {len(non_appearances)}")

print("\nMatch methods:")
print(fixtures["match_method"].value_counts().to_string())

print("\nDate discrepancies resolved:")
print(
    date_discrepancies.to_string(index=False)
    if len(date_discrepancies)
    else "None"
)

print("\nCandidate non-appearance fixtures:")
print(non_appearances.to_string(index=False))

print("\nUnmatched canonical appearances:")
if len(unmatched_apps):
    print(
        unmatched_apps[
            ["date", "opponent", "competition", "match_id"]
        ].to_string(index=False)
    )
else:
    print("None")
```

```python
# Collect Manchester United first-spell official fixtures
# 2003/04 through 2008/09 from the club's season pages.

import requests
from bs4 import BeautifulSoup
from io import StringIO

season_urls = {
    "2003-04": "https://en.wikipedia.org/wiki/2003%E2%80%9304_Manchester_United_F.C._season",
    "2004-05": "https://en.wikipedia.org/wiki/2004%E2%80%9305_Manchester_United_F.C._season",
    "2005-06": "https://en.wikipedia.org/wiki/2005%E2%80%9306_Manchester_United_F.C._season",
    "2006-07": "https://en.wikipedia.org/wiki/2006%E2%80%9307_Manchester_United_F.C._season",
    "2007-08": "https://en.wikipedia.org/wiki/2007%E2%80%9308_Manchester_United_F.C._season",
    "2008-09": "https://en.wikipedia.org/wiki/2008%E2%80%9309_Manchester_United_F.C._season",
}

# Only official competitions.
competition_map = {
    "FA Community Shield": "FA Community Shield",
    "Community Shield": "FA Community Shield",
    "Premier League": "Premier League",
    "FA Premier League": "Premier League",
    "FA Cup": "FA Cup",
    "League Cup": "League Cup",
    "UEFA Champions League": "Champions League",
    "UEFA Super Cup": "UEFA Super Cup",
    "FIFA Club World Cup": "FIFA Club World Cup",
    "Club World Cup": "FIFA Club World Cup",
}

headers = {
    "User-Agent": "Mozilla/5.0"
}


def flatten_columns(columns):
    """Convert pandas MultiIndex/complex table headers into simple names."""
    if isinstance(columns, pd.MultiIndex):
        cleaned = []
        for col in columns:
            parts = [
                str(x).strip()
                for x in col
                if str(x) != "nan" and not str(x).startswith("Unnamed")
            ]
            cleaned.append(" ".join(dict.fromkeys(parts)))
        return cleaned

    return [str(c).strip() for c in columns]


def find_column(columns, keywords):
    """Return first column containing any supplied keyword."""
    for col in columns:
        key = col.lower().strip()
        if any(word in key for word in keywords):
            return col
    return None


all_match_tables = []
table_audit = []

for season, url in season_urls.items():

    response = requests.get(url, headers=headers, timeout=30)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    current_h2 = None

    # Walk through headings and tables in page order.
    for element in soup.find_all(["h2", "table"]):

        if element.name == "h2":
            current_h2 = (
                element.get_text(" ", strip=True)
                .replace("[edit]", "")
                .strip()
            )
            continue

        if current_h2 not in competition_map:
            continue

        try:
            table = pd.read_html(StringIO(str(element)))[0]
        except Exception:
            continue

        table.columns = flatten_columns(table.columns)

        date_col = find_column(table.columns, ["date"])
        opponent_col = find_column(
            table.columns,
            ["opponent", "opponents", "opposition"]
        )
        result_col = find_column(
            table.columns,
            ["result", "score", "f–a", "f-a"]
        )
        venue_col = find_column(
            table.columns,
            ["h/a", "venue"]
        )

        # Fixture tables must contain at least date + opponent + result.
        if not all([date_col, opponent_col, result_col]):
            continue

        extracted = pd.DataFrame({
            "season": season,
            "date": table[date_col],
            "team": "Manchester United",
            "opponent": table[opponent_col],
            "competition": competition_map[current_h2],
            "venue": (
                table[venue_col]
                if venue_col is not None
                else pd.NA
            ),
            "raw_result": table[result_col],
            "source": url,
        })

        # Remove repeated header rows / invalid dates.
        extracted["date_parsed"] = pd.to_datetime(
            extracted["date"],
            errors="coerce",
            dayfirst=True
        )

        extracted = extracted[
            extracted["date_parsed"].notna()
        ].copy()

        if extracted.empty:
            continue

        extracted["date"] = extracted["date_parsed"]
        extracted.drop(columns="date_parsed", inplace=True)

        all_match_tables.append(extracted)

        table_audit.append({
            "season": season,
            "competition": competition_map[current_h2],
            "rows": len(extracted),
        })


man_utd_first_spell_fixtures = pd.concat(
    all_match_tables,
    ignore_index=True
)

# Tenure starts when Ronaldo joined United.
man_utd_first_spell_fixtures = man_utd_first_spell_fixtures[
    (man_utd_first_spell_fixtures["date"] >= "2003-08-12") &
    (man_utd_first_spell_fixtures["date"] <= "2009-06-30")
].copy()

man_utd_first_spell_fixtures = (
    man_utd_first_spell_fixtures
    .sort_values(["date", "competition", "opponent"])
    .reset_index(drop=True)
)

table_audit = pd.DataFrame(table_audit)

print("MANCHESTER UNITED FIRST SPELL — RAW OFFICIAL FIXTURE COLLECTION")
print("=" * 100)
print(f"Rows collected: {len(man_utd_first_spell_fixtures)}")
print(
    f"Date range: "
    f"{man_utd_first_spell_fixtures['date'].min().date()} "
    f"→ {man_utd_first_spell_fixtures['date'].max().date()}"
)

print("\nRows by season and competition:")
print(
    man_utd_first_spell_fixtures
    .groupby(["season", "competition"])
    .size()
    .unstack(fill_value=0)
    .to_string()
)

print("\nTotal by season:")
print(
    man_utd_first_spell_fixtures
    .groupby("season")
    .size()
    .to_string()
)

print("\nDuplicate candidate rows:")
duplicates = man_utd_first_spell_fixtures[
    man_utd_first_spell_fixtures.duplicated(
        ["date", "opponent", "competition"],
        keep=False
    )
]

print(
    duplicates[
        ["season", "date", "opponent", "competition", "raw_result"]
    ].to_string(index=False)
    if len(duplicates)
    else "None"
)

print("\nFirst 10 rows:")
print(
    man_utd_first_spell_fixtures[
        ["date", "opponent", "competition", "venue", "raw_result"]
    ].head(10).to_string(index=False)
)

print("\nLast 10 rows:")
print(
    man_utd_first_spell_fixtures[
        ["date", "opponent", "competition", "venue", "raw_result"]
    ].tail(10).to_string(index=False)
)
```

```python
# Repair missing H/A/N venues from the original season tables

venue_rows = []

for season, url in season_urls.items():

    response = requests.get(url, headers=headers, timeout=30)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")
    current_h2 = None

    for element in soup.find_all(["h2", "table"]):

        if element.name == "h2":
            current_h2 = (
                element.get_text(" ", strip=True)
                .replace("[edit]", "")
                .strip()
            )
            continue

        if current_h2 not in competition_map:
            continue

        try:
            table = pd.read_html(StringIO(str(element)))[0]
        except Exception:
            continue

        table.columns = flatten_columns(table.columns)

        date_col = find_column(table.columns, ["date"])
        opponent_col = find_column(
            table.columns,
            ["opponent", "opponents", "opposition"]
        )

        # Explicitly detect the H/A column.
        venue_col = next(
            (
                col for col in table.columns
                if col.lower().replace(" ", "") in {
                    "h/a",
                    "h-a",
                    "home/away",
                    "home-away",
                }
            ),
            None,
        )

        if date_col is None or opponent_col is None or venue_col is None:
            continue

        temp = pd.DataFrame({
            "season": season,
            "date": pd.to_datetime(
                table[date_col],
                errors="coerce",
                dayfirst=True,
            ),
            "opponent": table[opponent_col].astype(str).str.strip(),
            "competition": competition_map[current_h2],
            "recovered_venue": (
                table[venue_col]
                .astype(str)
                .str.strip()
                .str.upper()
            ),
        })

        temp = temp[temp["date"].notna()].copy()

        venue_rows.append(temp)


venue_lookup = pd.concat(
    venue_rows,
    ignore_index=True
)

# Remove exact duplicate lookup rows caused by split competition tables.
venue_lookup = venue_lookup.drop_duplicates(
    ["season", "date", "opponent", "competition"]
)

# Merge recovered venue onto the original 354-fixture table.
man_utd_first_spell_fixtures = (
    man_utd_first_spell_fixtures
    .merge(
        venue_lookup,
        on=["season", "date", "opponent", "competition"],
        how="left",
        validate="one_to_one",
    )
)

# Fill only values that were missing.
man_utd_first_spell_fixtures["venue"] = (
    man_utd_first_spell_fixtures["venue"]
    .fillna(man_utd_first_spell_fixtures["recovered_venue"])
)

man_utd_first_spell_fixtures.drop(
    columns="recovered_venue",
    inplace=True,
)

# Normalize and validate.
man_utd_first_spell_fixtures["venue"] = (
    man_utd_first_spell_fixtures["venue"]
    .astype("string")
    .str.strip()
    .str.upper()
)

valid_venues = {"H", "A", "N"}

invalid_venues = man_utd_first_spell_fixtures[
    man_utd_first_spell_fixtures["venue"].notna()
    & ~man_utd_first_spell_fixtures["venue"].isin(valid_venues)
]

print("MANCHESTER UNITED FIRST SPELL — VENUE REPAIR")
print("=" * 90)

print(f"Total fixtures: {len(man_utd_first_spell_fixtures)}")
print(
    f"Venue complete: "
    f"{man_utd_first_spell_fixtures['venue'].notna().sum()}"
)
print(
    f"Venue still missing: "
    f"{man_utd_first_spell_fixtures['venue'].isna().sum()}"
)
print(f"Invalid venue values: {len(invalid_venues)}")

print("\nFinal venue distribution:")
print(
    man_utd_first_spell_fixtures["venue"]
    .value_counts(dropna=False)
    .to_string()
)

print("\nVenue by season:")
print(
    pd.crosstab(
        man_utd_first_spell_fixtures["season"],
        man_utd_first_spell_fixtures["venue"],
        dropna=False,
    ).to_string()
)

if len(invalid_venues):
    print("\nInvalid rows:")
    print(
        invalid_venues[
            ["season", "date", "opponent", "competition", "venue"]
        ].to_string(index=False)
    )
```

```python
# Reconcile Manchester United 2003–2009 fixtures
# against Ronaldo's canonical first-spell appearances

import pandas as pd
import re
import unicodedata


# ------------------------------------------------------------
# 1. Extract Ronaldo's canonical Manchester United first spell
# ------------------------------------------------------------

canonical_dates = pd.to_datetime(df["date"])

man_utd_first_apps = df[
    (df["team"] == "Manchester United") &
    (canonical_dates >= "2003-08-12") &
    (canonical_dates <= "2009-06-30")
].copy()

man_utd_first_apps["date"] = pd.to_datetime(
    man_utd_first_apps["date"]
)

man_utd_first_apps = (
    man_utd_first_apps
    .sort_values("date")
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 2. Normalize text for reconciliation only
#    Original names remain untouched.
# ------------------------------------------------------------

def normalize_match_text(value):
    value = str(value).strip()

    value = unicodedata.normalize("NFKD", value)
    value = "".join(
        char for char in value
        if not unicodedata.combining(char)
    )

    value = value.lower()
    value = value.replace("&", "and")
    value = re.sub(r"[^a-z0-9]+", " ", value)
    value = re.sub(r"\s+", " ", value).strip()

    return value


# Known opponent naming differences between sources
opponent_aliases = {
    "internazionale": "inter milan",
    "inter": "inter milan",
    "as roma": "roma",
    "ac milan": "milan",
    "fc porto": "porto",
    "sporting cp": "sporting cp",
    "sporting lisbon": "sporting cp",
    "dynamo kyiv": "dynamo kyiv",
    "dynamo kiev": "dynamo kyiv",
    "aalborg bk": "aalborg",
    "alborg bk": "aalborg",
    "lille osc": "lille",
    "olympique lyon": "lyon",
    "olympique lyonnais": "lyon",
    "vfb stuttgart": "stuttgart",
    "fenerbahce": "fenerbahce",
    "sparta prague": "sparta prague",
    "benfica": "benfica",
    "villarreal": "villarreal",
    "celtic": "celtic",
    "debrecen": "debrecen",
    "gamba osaka": "gamba osaka",
    "ldu quito": "liga de quito",
}


def normalize_opponent(value):
    key = normalize_match_text(value)
    return opponent_aliases.get(key, key)


# Competition normalization is only for matching.
competition_aliases = {
    "champions league": "champions league",
    "uefa champions league": "champions league",

    "premier league": "premier league",
    "fa premier league": "premier league",

    "fa cup": "fa cup",

    "league cup": "league cup",
    "efl cup": "league cup",
    "carling cup": "league cup",

    "fa community shield": "community shield",
    "community shield": "community shield",

    "uefa super cup": "uefa super cup",

    "fifa club world cup": "club world cup",
    "club world cup": "club world cup",
}


def normalize_competition(value):
    key = normalize_match_text(value)
    return competition_aliases.get(key, key)


fixtures_mu = man_utd_first_spell_fixtures.copy()
apps_mu = man_utd_first_apps.copy()

fixtures_mu["opponent_key"] = (
    fixtures_mu["opponent"].map(normalize_opponent)
)
apps_mu["opponent_key"] = (
    apps_mu["opponent"].map(normalize_opponent)
)

fixtures_mu["competition_key"] = (
    fixtures_mu["competition"].map(normalize_competition)
)
apps_mu["competition_key"] = (
    apps_mu["competition"].map(normalize_competition)
)


# ------------------------------------------------------------
# 3. Prepare reconciliation fields
# ------------------------------------------------------------

fixtures_mu["appeared"] = 0
fixtures_mu["canonical_match_id"] = pd.NA
fixtures_mu["canonical_date"] = pd.NaT
fixtures_mu["match_method"] = "unmatched"

used_app_indices = set()


# ------------------------------------------------------------
# 4. Pass 1 — exact date + opponent + competition
# ------------------------------------------------------------

for fixture_idx, fixture in fixtures_mu.iterrows():

    candidates = apps_mu[
        (apps_mu["date"] == fixture["date"]) &
        (apps_mu["opponent_key"] == fixture["opponent_key"]) &
        (
            apps_mu["competition_key"]
            == fixture["competition_key"]
        )
    ]

    candidates = candidates[
        ~candidates.index.isin(used_app_indices)
    ]

    if len(candidates) == 1:

        app_idx = candidates.index[0]

        fixtures_mu.loc[fixture_idx, "appeared"] = 1
        fixtures_mu.loc[
            fixture_idx, "canonical_match_id"
        ] = apps_mu.loc[app_idx, "match_id"]

        fixtures_mu.loc[
            fixture_idx, "canonical_date"
        ] = apps_mu.loc[app_idx, "date"]

        fixtures_mu.loc[
            fixture_idx, "match_method"
        ] = "exact"

        used_app_indices.add(app_idx)


# ------------------------------------------------------------
# 5. Pass 2 — ±1 day source-date discrepancy
# ------------------------------------------------------------

for fixture_idx, fixture in fixtures_mu[
    fixtures_mu["match_method"] == "unmatched"
].iterrows():

    candidates = apps_mu[
        (apps_mu["opponent_key"] == fixture["opponent_key"]) &
        (
            apps_mu["competition_key"]
            == fixture["competition_key"]
        ) &
        (~apps_mu.index.isin(used_app_indices))
    ].copy()

    if candidates.empty:
        continue

    candidates["date_difference_days"] = (
        candidates["date"] - fixture["date"]
    ).abs().dt.days

    candidates = candidates[
        candidates["date_difference_days"] <= 1
    ]

    if len(candidates) == 1:

        app_idx = candidates.index[0]

        fixtures_mu.loc[fixture_idx, "appeared"] = 1

        fixtures_mu.loc[
            fixture_idx, "canonical_match_id"
        ] = apps_mu.loc[app_idx, "match_id"]

        fixtures_mu.loc[
            fixture_idx, "canonical_date"
        ] = apps_mu.loc[app_idx, "date"]

        fixtures_mu.loc[
            fixture_idx, "match_method"
        ] = "date_discrepancy"

        used_app_indices.add(app_idx)


# ------------------------------------------------------------
# 6. Audit results
# ------------------------------------------------------------

unmatched_apps_mu = apps_mu[
    ~apps_mu.index.isin(used_app_indices)
].copy()

candidate_absences_mu = fixtures_mu[
    fixtures_mu["appeared"] == 0
].copy()

date_discrepancies_mu = fixtures_mu[
    fixtures_mu["match_method"] == "date_discrepancy"
].copy()


print("MANCHESTER UNITED FIRST SPELL — RECONCILIATION")
print("=" * 100)

print(f"Official team fixtures: {len(fixtures_mu)}")
print(f"Canonical Ronaldo appearances: {len(apps_mu)}")
print(f"Successfully reconciled: {fixtures_mu['appeared'].sum()}")
print(f"Unmatched canonical appearances: {len(unmatched_apps_mu)}")
print(f"Candidate non-appearances: {len(candidate_absences_mu)}")

print("\nMatch methods:")
print(
    fixtures_mu["match_method"]
    .value_counts()
    .to_string()
)

print("\nReconciliation by season:")
print(
    fixtures_mu
    .groupby("season")
    .agg(
        fixtures=("date", "size"),
        appeared=("appeared", "sum"),
    )
    .assign(
        non_appearances=lambda x: (
            x["fixtures"] - x["appeared"]
        )
    )
    .to_string()
)

print("\nDate discrepancies:")
if len(date_discrepancies_mu):
    print(
        date_discrepancies_mu[
            [
                "date",
                "canonical_date",
                "opponent",
                "competition",
                "canonical_match_id",
            ]
        ].to_string(index=False)
    )
else:
    print("None")

print("\nUNMATCHED CANONICAL APPEARANCES:")
if len(unmatched_apps_mu):
    print(
        unmatched_apps_mu[
            [
                "date",
                "opponent",
                "competition",
                "venue",
                "match_id",
            ]
        ].to_string(index=False)
    )
else:
    print("None")

print("\nCandidate non-appearances by season:")
print(
    candidate_absences_mu
    .groupby("season")
    .size()
    .to_string()
)
```

```python
# Reconcile Manchester United 2003–2009 fixtures
# against Ronaldo's canonical first-spell appearances

import pandas as pd
import re
import unicodedata


# ------------------------------------------------------------
# 1. Extract Ronaldo's canonical Manchester United first spell
# ------------------------------------------------------------

canonical_dates = pd.to_datetime(df["date"])

man_utd_first_apps = df[
    (df["team"] == "Manchester United") &
    (canonical_dates >= "2003-08-12") &
    (canonical_dates <= "2009-06-30")
].copy()

man_utd_first_apps["date"] = pd.to_datetime(
    man_utd_first_apps["date"]
)

man_utd_first_apps = (
    man_utd_first_apps
    .sort_values("date")
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 2. Normalize text for reconciliation only
#    Original names remain untouched.
# ------------------------------------------------------------

def normalize_match_text(value):
    value = str(value).strip()

    value = unicodedata.normalize("NFKD", value)
    value = "".join(
        char for char in value
        if not unicodedata.combining(char)
    )

    value = value.lower()
    value = value.replace("&", "and")
    value = re.sub(r"[^a-z0-9]+", " ", value)
    value = re.sub(r"\s+", " ", value).strip()

    return value


# Known opponent naming differences between sources
opponent_aliases = {
    "internazionale": "inter milan",
    "inter": "inter milan",
    "as roma": "roma",
    "ac milan": "milan",
    "fc porto": "porto",
    "sporting cp": "sporting cp",
    "sporting lisbon": "sporting cp",
    "dynamo kyiv": "dynamo kyiv",
    "dynamo kiev": "dynamo kyiv",
    "aalborg bk": "aalborg",
    "alborg bk": "aalborg",
    "lille osc": "lille",
    "olympique lyon": "lyon",
    "olympique lyonnais": "lyon",
    "vfb stuttgart": "stuttgart",
    "fenerbahce": "fenerbahce",
    "sparta prague": "sparta prague",
    "benfica": "benfica",
    "villarreal": "villarreal",
    "celtic": "celtic",
    "debrecen": "debrecen",
    "gamba osaka": "gamba osaka",
    "ldu quito": "liga de quito",
}


def normalize_opponent(value):
    key = normalize_match_text(value)
    return opponent_aliases.get(key, key)


# Competition normalization is only for matching.
competition_aliases = {
    "champions league": "champions league",
    "uefa champions league": "champions league",

    "premier league": "premier league",
    "fa premier league": "premier league",

    "fa cup": "fa cup",

    "league cup": "league cup",
    "efl cup": "league cup",
    "carling cup": "league cup",

    "fa community shield": "community shield",
    "community shield": "community shield",

    "uefa super cup": "uefa super cup",

    "fifa club world cup": "club world cup",
    "club world cup": "club world cup",
}


def normalize_competition(value):
    key = normalize_match_text(value)
    return competition_aliases.get(key, key)


fixtures_mu = man_utd_first_spell_fixtures.copy()
apps_mu = man_utd_first_apps.copy()

fixtures_mu["opponent_key"] = (
    fixtures_mu["opponent"].map(normalize_opponent)
)
apps_mu["opponent_key"] = (
    apps_mu["opponent"].map(normalize_opponent)
)

fixtures_mu["competition_key"] = (
    fixtures_mu["competition"].map(normalize_competition)
)
apps_mu["competition_key"] = (
    apps_mu["competition"].map(normalize_competition)
)


# ------------------------------------------------------------
# 3. Prepare reconciliation fields
# ------------------------------------------------------------

fixtures_mu["appeared"] = 0
fixtures_mu["canonical_match_id"] = pd.NA
fixtures_mu["canonical_date"] = pd.NaT
fixtures_mu["match_method"] = "unmatched"

used_app_indices = set()


# ------------------------------------------------------------
# 4. Pass 1 — exact date + opponent + competition
# ------------------------------------------------------------

for fixture_idx, fixture in fixtures_mu.iterrows():

    candidates = apps_mu[
        (apps_mu["date"] == fixture["date"]) &
        (apps_mu["opponent_key"] == fixture["opponent_key"]) &
        (
            apps_mu["competition_key"]
            == fixture["competition_key"]
        )
    ]

    candidates = candidates[
        ~candidates.index.isin(used_app_indices)
    ]

    if len(candidates) == 1:

        app_idx = candidates.index[0]

        fixtures_mu.loc[fixture_idx, "appeared"] = 1
        fixtures_mu.loc[
            fixture_idx, "canonical_match_id"
        ] = apps_mu.loc[app_idx, "match_id"]

        fixtures_mu.loc[
            fixture_idx, "canonical_date"
        ] = apps_mu.loc[app_idx, "date"]

        fixtures_mu.loc[
            fixture_idx, "match_method"
        ] = "exact"

        used_app_indices.add(app_idx)


# ------------------------------------------------------------
# 5. Pass 2 — ±1 day source-date discrepancy
# ------------------------------------------------------------

for fixture_idx, fixture in fixtures_mu[
    fixtures_mu["match_method"] == "unmatched"
].iterrows():

    candidates = apps_mu[
        (apps_mu["opponent_key"] == fixture["opponent_key"]) &
        (
            apps_mu["competition_key"]
            == fixture["competition_key"]
        ) &
        (~apps_mu.index.isin(used_app_indices))
    ].copy()

    if candidates.empty:
        continue

    candidates["date_difference_days"] = (
        candidates["date"] - fixture["date"]
    ).abs().dt.days

    candidates = candidates[
        candidates["date_difference_days"] <= 1
    ]

    if len(candidates) == 1:

        app_idx = candidates.index[0]

        fixtures_mu.loc[fixture_idx, "appeared"] = 1

        fixtures_mu.loc[
            fixture_idx, "canonical_match_id"
        ] = apps_mu.loc[app_idx, "match_id"]

        fixtures_mu.loc[
            fixture_idx, "canonical_date"
        ] = apps_mu.loc[app_idx, "date"]

        fixtures_mu.loc[
            fixture_idx, "match_method"
        ] = "date_discrepancy"

        used_app_indices.add(app_idx)


# ------------------------------------------------------------
# 6. Audit results
# ------------------------------------------------------------

unmatched_apps_mu = apps_mu[
    ~apps_mu.index.isin(used_app_indices)
].copy()

candidate_absences_mu = fixtures_mu[
    fixtures_mu["appeared"] == 0
].copy()

date_discrepancies_mu = fixtures_mu[
    fixtures_mu["match_method"] == "date_discrepancy"
].copy()


print("MANCHESTER UNITED FIRST SPELL — RECONCILIATION")
print("=" * 100)

print(f"Official team fixtures: {len(fixtures_mu)}")
print(f"Canonical Ronaldo appearances: {len(apps_mu)}")
print(f"Successfully reconciled: {fixtures_mu['appeared'].sum()}")
print(f"Unmatched canonical appearances: {len(unmatched_apps_mu)}")
print(f"Candidate non-appearances: {len(candidate_absences_mu)}")

print("\nMatch methods:")
print(
    fixtures_mu["match_method"]
    .value_counts()
    .to_string()
)

print("\nReconciliation by season:")
print(
    fixtures_mu
    .groupby("season")
    .agg(
        fixtures=("date", "size"),
        appeared=("appeared", "sum"),
    )
    .assign(
        non_appearances=lambda x: (
            x["fixtures"] - x["appeared"]
        )
    )
    .to_string()
)

print("\nDate discrepancies:")
if len(date_discrepancies_mu):
    print(
        date_discrepancies_mu[
            [
                "date",
                "canonical_date",
                "opponent",
                "competition",
                "canonical_match_id",
            ]
        ].to_string(index=False)
    )
else:
    print("None")

print("\nUNMATCHED CANONICAL APPEARANCES:")
if len(unmatched_apps_mu):
    print(
        unmatched_apps_mu[
            [
                "date",
                "opponent",
                "competition",
                "venue",
                "match_id",
            ]
        ].to_string(index=False)
    )
else:
    print("None")

print("\nCandidate non-appearances by season:")
print(
    candidate_absences_mu
    .groupby("season")
    .size()
    .to_string()
)
```

```python
# Collect all official Real Madrid fixtures during Ronaldo's tenure
# using Wikipedia football-box (vevent) structures.



# ------------------------------------------------------------
# Real Madrid season pages
# ------------------------------------------------------------

real_madrid_seasons = {
    "2009-10": "https://en.wikipedia.org/wiki/2009%E2%80%9310_Real_Madrid_CF_season",
    "2010-11": "https://en.wikipedia.org/wiki/2010%E2%80%9311_Real_Madrid_CF_season",
    "2011-12": "https://en.wikipedia.org/wiki/2011%E2%80%9312_Real_Madrid_CF_season",
    "2012-13": "https://en.wikipedia.org/wiki/2012%E2%80%9313_Real_Madrid_CF_season",
    "2013-14": "https://en.wikipedia.org/wiki/2013%E2%80%9314_Real_Madrid_CF_season",
    "2014-15": "https://en.wikipedia.org/wiki/2014%E2%80%9315_Real_Madrid_CF_season",
    "2015-16": "https://en.wikipedia.org/wiki/2015%E2%80%9316_Real_Madrid_CF_season",
    "2016-17": "https://en.wikipedia.org/wiki/2016%E2%80%9317_Real_Madrid_CF_season",
    "2017-18": "https://en.wikipedia.org/wiki/2017%E2%80%9318_Real_Madrid_CF_season",
}


# ------------------------------------------------------------
# Helpers
# ------------------------------------------------------------

def clean_text(value):
    value = unicodedata.normalize("NFKD", str(value))
    value = "".join(
        ch for ch in value
        if not unicodedata.combining(ch)
    )
    value = re.sub(r"\[[^\]]+\]", "", value)
    value = re.sub(r"\s+", " ", value)
    return value.strip()


def competition_from_box(box):
    """
    Walk backward through section headings until we find
    one of Real Madrid's official competitions.
    """

    allowed = {
        "la liga": "La Liga",
        "copa del rey": "Copa del Rey",
        "uefa champions league": "Champions League",
        "champions league": "Champions League",
        "supercopa de espana": "Supercopa de España",
        "uefa super cup": "UEFA Super Cup",
        "fifa club world cup": "FIFA Club World Cup",
        "club world cup": "FIFA Club World Cup",
    }

    for heading in box.find_all_previous(
        ["h2", "h3", "h4", "h5", "h6"]
    ):
        text = clean_text(
            heading.get_text(" ", strip=True)
        ).lower()

        if text in allowed:
            return allowed[text]

    return None


def extract_date_from_box(box):
    """
    Extract the first full calendar date from the match box.
    """

    text = box.get_text(" ", strip=True)

    month_pattern = (
        r"(?:January|February|March|April|May|June|"
        r"July|August|September|October|November|December)"
    )

    pattern = (
        rf"\b\d{{1,2}}\s+{month_pattern}\s+\d{{4}}\b"
    )

    match = re.search(pattern, text)

    if not match:
        return pd.NaT

    return pd.to_datetime(
        match.group(0),
        dayfirst=True,
        errors="coerce",
    )


def extract_score_from_box(box):
    """
    Extract the match score such as 3–2 or 1–1.
    """

    text = box.get_text(" ", strip=True)

    # Handles both en dash and ordinary hyphen
    match = re.search(
        r"\b(\d{1,2})\s*[–-]\s*(\d{1,2})\b",
        text,
    )

    if not match:
        return None

    return f"{match.group(1)}-{match.group(2)}"


def extract_teams_from_box(box):
    """
    Football-box attendees appear in home-team, away-team order.
    """

    attendee_cells = box.select(
        "td.vcard.attendee"
    )

    if len(attendee_cells) < 2:
        return None, None

    home_team = attendee_cells[0].get_text(
        " ",
        strip=True,
    )

    away_team = attendee_cells[1].get_text(
        " ",
        strip=True,
    )

    return (
        clean_text(home_team),
        clean_text(away_team),
    )


def madrid_opponent_and_venue(home_team, away_team):
    """
    Derive opponent and provisional H/A venue from team order.

    Neutral-site corrections, if any, will be audited separately.
    """

    home_key = clean_text(home_team).lower()
    away_key = clean_text(away_team).lower()

    madrid_names = {
        "real madrid",
        "real madrid cf",
    }

    if home_key in madrid_names:
        return away_team, "H"

    if away_key in madrid_names:
        return home_team, "A"

    return None, None


# ------------------------------------------------------------
# Collect one season
# ------------------------------------------------------------

def collect_real_madrid_season(season, url):

    response = requests.get(
        url,
        headers={
            "User-Agent": "Mozilla/5.0"
        },
        timeout=30,
    )
    response.raise_for_status()

    soup = BeautifulSoup(
        response.text,
        "html.parser",
    )

    rows = []

    # Only outer DIV vevents.
    # This avoids also grabbing the nested football-box table.
    match_boxes = soup.select(
        "div.vevent"
    )

    for box in match_boxes:

        competition = competition_from_box(box)

        # Ignore friendlies and anything outside the official
        # competition list.
        if competition is None:
            continue

        date = extract_date_from_box(box)

        home_team, away_team = (
            extract_teams_from_box(box)
        )

        if (
            pd.isna(date)
            or home_team is None
            or away_team is None
        ):
            continue

        opponent, venue = madrid_opponent_and_venue(
            home_team,
            away_team,
        )

        # The box must actually involve Real Madrid.
        if opponent is None:
            continue

        score = extract_score_from_box(box)

        rows.append({
            "season": season,
            "date": date,
            "team": "Real Madrid",
            "opponent": opponent,
            "competition": competition,
            "venue": venue,
            "home_team": home_team,
            "away_team": away_team,
            "raw_result": score,
            "source": url,
        })

    return pd.DataFrame(rows)


# ------------------------------------------------------------
# Collect the entire nine-season tenure
# ------------------------------------------------------------

real_madrid_parts = []

for season, url in real_madrid_seasons.items():

    season_df = collect_real_madrid_season(
        season,
        url,
    )

    print(
        f"{season}: "
        f"{len(season_df)} official fixtures collected"
    )

    real_madrid_parts.append(
        season_df
    )


# ------------------------------------------------------------
# Combine
# ------------------------------------------------------------

non_empty_parts = [
    x for x in real_madrid_parts
    if not x.empty
]

if not non_empty_parts:
    raise ValueError(
        "No Real Madrid fixtures were collected."
    )

real_madrid_fixtures = (
    pd.concat(
        non_empty_parts,
        ignore_index=True,
    )
    .sort_values(
        ["date", "competition", "opponent"]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Restrict to Ronaldo's Madrid tenure
# ------------------------------------------------------------

real_madrid_fixtures = (
    real_madrid_fixtures[
        real_madrid_fixtures["date"].between(
            pd.Timestamp("2009-07-01"),
            pd.Timestamp("2018-07-09"),
        )
    ]
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Duplicate audit
# ------------------------------------------------------------

duplicate_mask = (
    real_madrid_fixtures.duplicated(
        subset=[
            "date",
            "opponent",
            "competition",
        ],
        keep=False,
    )
)


# ------------------------------------------------------------
# Final collection audit
# ------------------------------------------------------------

print(
    "\nREAL MADRID — FULL TENURE "
    "FIXTURE COLLECTION"
)
print("=" * 100)

print(
    f"Total official fixtures collected: "
    f"{len(real_madrid_fixtures)}"
)

print(
    f"Date range: "
    f"{real_madrid_fixtures['date'].min().date()} "
    f"→ "
    f"{real_madrid_fixtures['date'].max().date()}"
)

print(
    f"Duplicate candidate rows: "
    f"{duplicate_mask.sum()}"
)

print(
    f"Missing dates: "
    f"{real_madrid_fixtures['date'].isna().sum()}"
)

print(
    f"Missing opponents: "
    f"{real_madrid_fixtures['opponent'].isna().sum()}"
)

print(
    f"Missing results: "
    f"{real_madrid_fixtures['raw_result'].isna().sum()}"
)

print(
    f"Missing venues: "
    f"{real_madrid_fixtures['venue'].isna().sum()}"
)


print("\nFixtures by season:")
print(
    real_madrid_fixtures[
        "season"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)


print("\nFixtures by competition:")
print(
    real_madrid_fixtures[
        "competition"
    ]
    .value_counts()
    .to_string()
)


print("\nSeason × competition:")
print(
    pd.crosstab(
        real_madrid_fixtures[
            "season"
        ],
        real_madrid_fixtures[
            "competition"
        ],
    ).to_string()
)


print("\nProvisional venue distribution:")
print(
    real_madrid_fixtures[
        "venue"
    ]
    .value_counts(
        dropna=False
    )
    .to_string()
)


print("\nFirst 5 fixtures:")
print(
    real_madrid_fixtures[
        [
            "date",
            "opponent",
            "competition",
            "venue",
            "raw_result",
        ]
    ]
    .head()
    .to_string(
        index=False
    )
)


print("\nLast 5 fixtures:")
print(
    real_madrid_fixtures[
        [
            "date",
            "opponent",
            "competition",
            "venue",
            "raw_result",
        ]
    ]
    .tail()
    .to_string(
        index=False
    )
)
```

```python
# Rebuild Real Madrid fixture collection with proper section boundaries




# ------------------------------------------------------------
# Real Madrid season pages
# ------------------------------------------------------------

real_madrid_seasons = {
    "2009-10": "https://en.wikipedia.org/wiki/2009%E2%80%9310_Real_Madrid_CF_season",
    "2010-11": "https://en.wikipedia.org/wiki/2010%E2%80%9311_Real_Madrid_CF_season",
    "2011-12": "https://en.wikipedia.org/wiki/2011%E2%80%9312_Real_Madrid_CF_season",
    "2012-13": "https://en.wikipedia.org/wiki/2012%E2%80%9313_Real_Madrid_CF_season",
    "2013-14": "https://en.wikipedia.org/wiki/2013%E2%80%9314_Real_Madrid_CF_season",
    "2014-15": "https://en.wikipedia.org/wiki/2014%E2%80%9315_Real_Madrid_CF_season",
    "2015-16": "https://en.wikipedia.org/wiki/2015%E2%80%9316_Real_Madrid_CF_season",
    "2016-17": "https://en.wikipedia.org/wiki/2016%E2%80%9317_Real_Madrid_CF_season",
    "2017-18": "https://en.wikipedia.org/wiki/2017%E2%80%9318_Real_Madrid_CF_season",
}


# ------------------------------------------------------------
# Helpers
# ------------------------------------------------------------

def clean_text(value):
    value = unicodedata.normalize("NFKD", str(value))
    value = "".join(
        ch for ch in value
        if not unicodedata.combining(ch)
    )
    value = re.sub(r"\[[^\]]+\]", "", value)
    value = re.sub(r"\s+", " ", value)
    return value.strip()


def competition_from_box(box):
    """
    Determine whether the match is competitive and, if so,
    return the competition name.

    Important:
    Matches under the H2 section 'Pre-season and friendlies'
    are rejected before looking at H3 headings.
    """

    nearest_h2 = box.find_previous("h2")

    if nearest_h2 is not None:
        h2_text = clean_text(
            nearest_h2.get_text(" ", strip=True)
        ).lower()

        if (
            "pre-season" in h2_text
            or "preseason" in h2_text
            or "friendlies" in h2_text
            or "friendly" in h2_text
        ):
            return None

    allowed_h3 = {
        "la liga": "La Liga",
        "copa del rey": "Copa del Rey",
        "uefa champions league": "Champions League",
        "champions league": "Champions League",
        "supercopa de espana": "Supercopa de España",
        "uefa super cup": "UEFA Super Cup",
        "fifa club world cup": "FIFA Club World Cup",
        "club world cup": "FIFA Club World Cup",
    }

    # Nearest H3 competition heading
    for heading in box.find_all_previous("h3"):

        text = clean_text(
            heading.get_text(" ", strip=True)
        ).lower()

        if text in allowed_h3:
            return allowed_h3[text]

        # Stop if we've crossed out of the competition section
        previous_h2 = heading.find_previous("h2")

        if previous_h2 is not None:
            previous_h2_text = clean_text(
                previous_h2.get_text(" ", strip=True)
            ).lower()

            if (
                "pre-season" in previous_h2_text
                or "friendlies" in previous_h2_text
            ):
                break

    return None


def extract_date_from_box(box):

    text = box.get_text(" ", strip=True)

    month_pattern = (
        r"(?:January|February|March|April|May|June|"
        r"July|August|September|October|November|December)"
    )

    match = re.search(
        rf"\b\d{{1,2}}\s+{month_pattern}\s+\d{{4}}\b",
        text,
    )

    if not match:
        return pd.NaT

    return pd.to_datetime(
        match.group(0),
        dayfirst=True,
        errors="coerce",
    )


def extract_score_from_box(box):

    text = box.get_text(" ", strip=True)

    match = re.search(
        r"\b(\d{1,2})\s*[–-]\s*(\d{1,2})\b",
        text,
    )

    if not match:
        return None

    return f"{match.group(1)}-{match.group(2)}"


def extract_teams_from_box(box):

    attendee_cells = box.select(
        "td.vcard.attendee"
    )

    if len(attendee_cells) < 2:
        return None, None

    home_team = clean_text(
        attendee_cells[0].get_text(
            " ",
            strip=True,
        )
    )

    away_team = clean_text(
        attendee_cells[1].get_text(
            " ",
            strip=True,
        )
    )

    return home_team, away_team


def madrid_opponent_and_venue(home_team, away_team):

    home_key = clean_text(home_team).lower()
    away_key = clean_text(away_team).lower()

    madrid_names = {
        "real madrid",
        "real madrid cf",
    }

    if home_key in madrid_names:
        return away_team, "H"

    if away_key in madrid_names:
        return home_team, "A"

    return None, None


# ------------------------------------------------------------
# Season collector
# ------------------------------------------------------------

def collect_real_madrid_season(season, url):

    response = requests.get(
        url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=30,
    )
    response.raise_for_status()

    soup = BeautifulSoup(
        response.text,
        "html.parser",
    )

    rows = []

    for box in soup.select("div.vevent"):

        competition = competition_from_box(box)

        if competition is None:
            continue

        date = extract_date_from_box(box)

        home_team, away_team = (
            extract_teams_from_box(box)
        )

        if (
            pd.isna(date)
            or home_team is None
            or away_team is None
        ):
            continue

        opponent, venue = madrid_opponent_and_venue(
            home_team,
            away_team,
        )

        if opponent is None:
            continue

        rows.append({
            "season": season,
            "date": date,
            "team": "Real Madrid",
            "opponent": opponent,
            "competition": competition,
            "venue": venue,
            "home_team": home_team,
            "away_team": away_team,
            "raw_result": extract_score_from_box(box),
            "source": url,
        })

    return pd.DataFrame(rows)


# ------------------------------------------------------------
# Collect all nine seasons
# ------------------------------------------------------------

real_madrid_parts = []

for season, url in real_madrid_seasons.items():

    season_df = collect_real_madrid_season(
        season,
        url,
    )

    print(
        f"{season}: "
        f"{len(season_df)} official fixtures collected"
    )

    real_madrid_parts.append(
        season_df
    )


real_madrid_fixtures = (
    pd.concat(
        [
            part
            for part in real_madrid_parts
            if not part.empty
        ],
        ignore_index=True,
    )
    .sort_values(
        ["date", "competition", "opponent"]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Strict Ronaldo tenure boundary
# ------------------------------------------------------------

real_madrid_fixtures = (
    real_madrid_fixtures[
        real_madrid_fixtures["date"].between(
            pd.Timestamp("2009-07-01"),
            pd.Timestamp("2018-07-09"),
        )
    ]
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Audit
# ------------------------------------------------------------

duplicate_mask = real_madrid_fixtures.duplicated(
    subset=[
        "date",
        "opponent",
        "competition",
    ],
    keep=False,
)


print(
    "\nREAL MADRID — CLEAN COMPETITIVE FIXTURE COLLECTION"
)
print("=" * 100)

print(
    f"Total official fixtures collected: "
    f"{len(real_madrid_fixtures)}"
)

print(
    f"Date range: "
    f"{real_madrid_fixtures['date'].min().date()} "
    f"→ "
    f"{real_madrid_fixtures['date'].max().date()}"
)

print(
    f"Duplicate candidate rows: "
    f"{duplicate_mask.sum()}"
)

print(
    f"Missing results: "
    f"{real_madrid_fixtures['raw_result'].isna().sum()}"
)

print("\nFixtures by season:")
print(
    real_madrid_fixtures[
        "season"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\nFixtures by competition:")
print(
    real_madrid_fixtures[
        "competition"
    ]
    .value_counts()
    .to_string()
)

print("\nSeason × competition:")
print(
    pd.crosstab(
        real_madrid_fixtures["season"],
        real_madrid_fixtures["competition"],
    ).to_string()
)

print("\nFirst 10 fixtures:")
print(
    real_madrid_fixtures[
        [
            "date",
            "opponent",
            "competition",
            "venue",
            "raw_result",
        ]
    ]
    .head(10)
    .to_string(index=False)
)
```

```python
# Finalize Real Madrid reconciliation and create appeared labels

# ------------------------------------------------------------
# Start from the clean 519-fixture universe
# ------------------------------------------------------------

madrid_appearance_table = (
    real_madrid_fixtures
    .copy()
    .reset_index(drop=True)
)

madrid_appearance_table["appeared"] = 0
madrid_appearance_table["match_status"] = "did_not_appear"
madrid_appearance_table["canonical_match_id"] = pd.NA


# ------------------------------------------------------------
# Rebuild canonical Real Madrid appearances
# ------------------------------------------------------------

madrid_apps = (
    canonical[
        canonical["team"].eq("Real Madrid")
    ]
    .copy()
    .sort_values("date")
    .reset_index(drop=True)
)

madrid_apps["opponent_key"] = (
    madrid_apps["opponent"]
    .map(normalize_opponent)
)

madrid_apps["competition_key"] = (
    madrid_apps["competition"]
    .map(normalize_competition)
)


madrid_appearance_table["opponent_key"] = (
    madrid_appearance_table["opponent"]
    .map(normalize_opponent)
)

madrid_appearance_table["competition_key"] = (
    madrid_appearance_table["competition"]
    .map(normalize_competition)
)


# ------------------------------------------------------------
# 1. Mark exact canonical appearance matches
# ------------------------------------------------------------

appearance_lookup = madrid_apps[
    [
        "match_id",
        "date",
        "opponent_key",
        "competition_key",
    ]
].copy()


exact_merge = (
    madrid_appearance_table
    .reset_index()
    .rename(columns={"index": "fixture_index"})
    .merge(
        appearance_lookup,
        on=[
            "date",
            "opponent_key",
            "competition_key",
        ],
        how="left",
        validate="one_to_one",
    )
)


exact_mask = exact_merge["match_id"].notna()

exact_fixture_indices = (
    exact_merge.loc[
        exact_mask,
        "fixture_index"
    ]
    .astype(int)
)


madrid_appearance_table.loc[
    exact_fixture_indices,
    "appeared"
] = 1

madrid_appearance_table.loc[
    exact_fixture_indices,
    "match_status"
] = "appeared_exact"

madrid_appearance_table.loc[
    exact_fixture_indices,
    "canonical_match_id"
] = (
    exact_merge.loc[
        exact_mask,
        "match_id"
    ]
    .values
)


# ------------------------------------------------------------
# 2. Explicitly reconcile the two verified date discrepancies
# ------------------------------------------------------------

date_discrepancies = [
    {
        "canonical_date": pd.Timestamp("2009-12-19"),
        "fixture_date": pd.Timestamp("2009-12-20"),
        "canonical_opponent": "Real Zaragoza",
        "fixture_opponent": "Zaragoza",
        "competition": "La Liga",
    },
    {
        "canonical_date": pd.Timestamp("2010-01-16"),
        "fixture_date": pd.Timestamp("2010-01-17"),
        "canonical_opponent": "Athletic Club",
        "fixture_opponent": "Athletic Bilbao",
        "competition": "La Liga",
    },
]


for discrepancy in date_discrepancies:

    canonical_row = madrid_apps[
        (madrid_apps["date"] == discrepancy["canonical_date"])
        &
        (
            madrid_apps["opponent_key"]
            == normalize_opponent(
                discrepancy["canonical_opponent"]
            )
        )
        &
        (
            madrid_apps["competition_key"]
            == normalize_competition(
                discrepancy["competition"]
            )
        )
    ]

    fixture_mask = (
        (madrid_appearance_table["date"] == discrepancy["fixture_date"])
        &
        (
            madrid_appearance_table["opponent_key"]
            == normalize_opponent(
                discrepancy["fixture_opponent"]
            )
        )
        &
        (
            madrid_appearance_table["competition_key"]
            == normalize_competition(
                discrepancy["competition"]
            )
        )
    )

    print(
        f"{discrepancy['canonical_date'].date()} "
        f"{discrepancy['canonical_opponent']} "
        f"→ "
        f"{discrepancy['fixture_date'].date()} "
        f"{discrepancy['fixture_opponent']}"
    )

    print(
        f"  canonical rows found: "
        f"{len(canonical_row)}"
    )

    print(
        f"  fixture rows found: "
        f"{fixture_mask.sum()}"
    )

    if len(canonical_row) != 1:
        raise ValueError(
            "Expected exactly one canonical row."
        )

    if fixture_mask.sum() != 1:
        raise ValueError(
            "Expected exactly one fixture row."
        )

    fixture_index = (
        madrid_appearance_table[
            fixture_mask
        ].index[0]
    )

    madrid_appearance_table.loc[
        fixture_index,
        "appeared"
    ] = 1

    madrid_appearance_table.loc[
        fixture_index,
        "match_status"
    ] = "appeared_date_reconciled"

    madrid_appearance_table.loc[
        fixture_index,
        "canonical_match_id"
    ] = canonical_row.iloc[0]["match_id"]


# ------------------------------------------------------------
# 3. Final integrity audit
# ------------------------------------------------------------

appeared_count = int(
    madrid_appearance_table["appeared"].sum()
)

nonappearance_count = int(
    (
        madrid_appearance_table["appeared"] == 0
    ).sum()
)

linked_match_ids = (
    madrid_appearance_table[
        madrid_appearance_table["appeared"] == 1
    ]["canonical_match_id"]
)


print("\nREAL MADRID — FINAL APPEARANCE LABEL AUDIT")
print("=" * 100)

print(
    f"Official fixtures: "
    f"{len(madrid_appearance_table)}"
)

print(
    f"Appeared = 1: "
    f"{appeared_count}"
)

print(
    f"Appeared = 0: "
    f"{nonappearance_count}"
)

print(
    f"Unique linked canonical match IDs: "
    f"{linked_match_ids.nunique()}"
)

print(
    f"Missing canonical IDs among appearances: "
    f"{linked_match_ids.isna().sum()}"
)

print("\nMatch status counts:")
print(
    madrid_appearance_table[
        "match_status"
    ]
    .value_counts()
    .to_string()
)


# ------------------------------------------------------------
# 4. Appearance rate by season
# ------------------------------------------------------------

print("\nAppearance rate by season:")
season_summary = (
    madrid_appearance_table
    .groupby("season")
    .agg(
        fixtures=("appeared", "size"),
        appearances=("appeared", "sum"),
    )
)

season_summary["non_appearances"] = (
    season_summary["fixtures"]
    - season_summary["appearances"]
)

season_summary["appearance_rate"] = (
    season_summary["appearances"]
    / season_summary["fixtures"]
)

print(
    season_summary.to_string()
)


# ------------------------------------------------------------
# 5. Save Real Madrid appearance dataset
# ------------------------------------------------------------

output_path = (
    "../data/appearance/"
    "real_madrid_2009_2018_appearance.csv"
)

madrid_appearance_table.to_csv(
    output_path,
    index=False,
)

print(
    f"\nSaved: {output_path}"
)
```

```python
# Collect Juventus official fixtures across Ronaldo's full tenure


# ------------------------------------------------------------
# Juventus season pages
# ------------------------------------------------------------

juventus_seasons = {
    "2018-19": (
        "https://en.wikipedia.org/wiki/"
        "2018%E2%80%9319_Juventus_FC_season"
    ),
    "2019-20": (
        "https://en.wikipedia.org/wiki/"
        "2019%E2%80%9320_Juventus_FC_season"
    ),
    "2020-21": (
        "https://en.wikipedia.org/wiki/"
        "2020%E2%80%9321_Juventus_FC_season"
    ),
}


# ------------------------------------------------------------
# Text normalization
# ------------------------------------------------------------

def clean_text(value):

    value = unicodedata.normalize(
        "NFKD",
        str(value),
    )

    value = "".join(
        ch for ch in value
        if not unicodedata.combining(ch)
    )

    value = re.sub(
        r"\[[^\]]+\]",
        "",
        value,
    )

    value = re.sub(
        r"\s+",
        " ",
        value,
    )

    return value.strip()


# ------------------------------------------------------------
# Competition detection
# ------------------------------------------------------------

def juventus_competition_from_box(box):

    # First protect against friendlies.
    nearest_h2 = box.find_previous("h2")

    if nearest_h2 is not None:

        h2_text = clean_text(
            nearest_h2.get_text(
                " ",
                strip=True,
            )
        ).lower()

        if (
            "pre-season" in h2_text
            or "preseason" in h2_text
            or "friendlies" in h2_text
            or "friendly" in h2_text
        ):
            return None


    allowed = {
        "serie a": "Serie A",

        "coppa italia": "Coppa Italia",

        "uefa champions league":
            "Champions League",

        "champions league":
            "Champions League",

        "supercoppa italiana":
            "Supercoppa Italiana",

        "supercoppa":
            "Supercoppa Italiana",
    }


    # Find nearest valid H3 competition heading.
    for heading in box.find_all_previous("h3"):

        text = clean_text(
            heading.get_text(
                " ",
                strip=True,
            )
        ).lower()

        if text in allowed:
            return allowed[text]

    return None


# ------------------------------------------------------------
# Date
# ------------------------------------------------------------

def extract_date_from_box(box):

    text = box.get_text(
        " ",
        strip=True,
    )

    month_pattern = (
        r"(?:January|February|March|April|May|June|"
        r"July|August|September|October|November|December)"
    )

    match = re.search(
        rf"\b\d{{1,2}}\s+{month_pattern}\s+\d{{4}}\b",
        text,
    )

    if not match:
        return pd.NaT

    return pd.to_datetime(
        match.group(0),
        dayfirst=True,
        errors="coerce",
    )


# ------------------------------------------------------------
# Score
# ------------------------------------------------------------

def extract_score_from_box(box):

    text = box.get_text(
        " ",
        strip=True,
    )

    match = re.search(
        r"\b(\d{1,2})\s*[–-]\s*(\d{1,2})\b",
        text,
    )

    if not match:
        return None

    return (
        f"{match.group(1)}-"
        f"{match.group(2)}"
    )


# ------------------------------------------------------------
# Teams
# ------------------------------------------------------------

def extract_teams_from_box(box):

    attendee_cells = box.select(
        "td.vcard.attendee"
    )

    if len(attendee_cells) < 2:
        return None, None

    home_team = clean_text(
        attendee_cells[0].get_text(
            " ",
            strip=True,
        )
    )

    away_team = clean_text(
        attendee_cells[1].get_text(
            " ",
            strip=True,
        )
    )

    return home_team, away_team


# ------------------------------------------------------------
# Opponent + provisional venue
# ------------------------------------------------------------

def juventus_opponent_and_venue(
    home_team,
    away_team,
):

    home_key = clean_text(
        home_team
    ).lower()

    away_key = clean_text(
        away_team
    ).lower()

    juventus_names = {
        "juventus",
        "juventus fc",
    }

    if home_key in juventus_names:
        return away_team, "H"

    if away_key in juventus_names:
        return home_team, "A"

    return None, None


# ------------------------------------------------------------
# One-season collector
# ------------------------------------------------------------

def collect_juventus_season(
    season,
    url,
):

    response = requests.get(
        url,
        headers={
            "User-Agent": "Mozilla/5.0"
        },
        timeout=30,
    )

    response.raise_for_status()

    soup = BeautifulSoup(
        response.text,
        "html.parser",
    )

    rows = []

    for box in soup.select("div.vevent"):

        competition = (
            juventus_competition_from_box(
                box
            )
        )

        if competition is None:
            continue

        date = extract_date_from_box(
            box
        )

        home_team, away_team = (
            extract_teams_from_box(
                box
            )
        )

        if (
            pd.isna(date)
            or home_team is None
            or away_team is None
        ):
            continue

        opponent, venue = (
            juventus_opponent_and_venue(
                home_team,
                away_team,
            )
        )

        if opponent is None:
            continue

        rows.append({
            "season": season,
            "date": date,
            "team": "Juventus",
            "opponent": opponent,
            "competition": competition,
            "venue": venue,
            "home_team": home_team,
            "away_team": away_team,
            "raw_result":
                extract_score_from_box(box),
            "source": url,
        })

    return pd.DataFrame(
        rows
    )


# ------------------------------------------------------------
# Collect all three seasons
# ------------------------------------------------------------

juventus_parts = []

for season, url in juventus_seasons.items():

    season_df = (
        collect_juventus_season(
            season,
            url,
        )
    )

    print(
        f"{season}: "
        f"{len(season_df)} "
        f"official fixtures collected"
    )

    juventus_parts.append(
        season_df
    )


juventus_fixtures = (
    pd.concat(
        [
            part
            for part in juventus_parts
            if not part.empty
        ],
        ignore_index=True,
    )
    .sort_values(
        [
            "date",
            "competition",
            "opponent",
        ]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Ronaldo Juventus tenure boundary
# ------------------------------------------------------------

juventus_fixtures = (
    juventus_fixtures[
        juventus_fixtures[
            "date"
        ].between(
            pd.Timestamp(
                "2018-07-10"
            ),
            pd.Timestamp(
                "2021-08-26"
            ),
        )
    ]
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Integrity checks
# ------------------------------------------------------------

duplicate_mask = (
    juventus_fixtures.duplicated(
        subset=[
            "date",
            "opponent",
            "competition",
        ],
        keep=False,
    )
)


print(
    "\nJUVENTUS — COMPETITIVE FIXTURE COLLECTION"
)
print("=" * 100)

print(
    f"Total official fixtures collected: "
    f"{len(juventus_fixtures)}"
)

print(
    f"Date range: "
    f"{juventus_fixtures['date'].min().date()} "
    f"→ "
    f"{juventus_fixtures['date'].max().date()}"
)

print(
    f"Duplicate candidate rows: "
    f"{duplicate_mask.sum()}"
)

print(
    f"Missing dates: "
    f"{juventus_fixtures['date'].isna().sum()}"
)

print(
    f"Missing opponents: "
    f"{juventus_fixtures['opponent'].isna().sum()}"
)

print(
    f"Missing results: "
    f"{juventus_fixtures['raw_result'].isna().sum()}"
)


print("\nFixtures by season:")
print(
    juventus_fixtures[
        "season"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)


print("\nFixtures by competition:")
print(
    juventus_fixtures[
        "competition"
    ]
    .value_counts()
    .to_string()
)


print("\nSeason × competition:")
print(
    pd.crosstab(
        juventus_fixtures[
            "season"
        ],
        juventus_fixtures[
            "competition"
        ],
    ).to_string()
)


print("\nFirst 5 fixtures:")
print(
    juventus_fixtures[
        [
            "date",
            "opponent",
            "competition",
            "venue",
            "raw_result",
        ]
    ]
    .head()
    .to_string(index=False)
)


print("\nLast 5 fixtures:")
print(
    juventus_fixtures[
        [
            "date",
            "opponent",
            "competition",
            "venue",
            "raw_result",
        ]
    ]
    .tail()
    .to_string(index=False)
)
```

```python
# Collect Manchester United second-spell official fixtures
# from the actual wikitable structure.

man_utd_second_seasons = {
    "2021-22": (
        "https://en.wikipedia.org/wiki/"
        "2021%E2%80%9322_Manchester_United_F.C._season"
    ),
    "2022-23": (
        "https://en.wikipedia.org/wiki/"
        "2022%E2%80%9323_Manchester_United_F.C._season"
    ),
}


competition_map = {
    "premier league": "Premier League",
    "fa cup": "FA Cup",
    "efl cup": "League Cup",
    "uefa champions league": "Champions League",
    "uefa europa league": "Europa League",
}


def collect_man_utd_second_season(season, url):

    response = requests.get(
        url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=30,
    )

    response.raise_for_status()

    soup = BeautifulSoup(
        response.text,
        "html.parser",
    )

    fixtures = []

    for table in soup.find_all("table", class_="wikitable"):

        # Competition is represented by the nearest H2.
        h2 = table.find_previous("h2")

        if h2 is None:
            continue

        h2_text = clean_text(
            h2.get_text(" ", strip=True)
        ).lower()

        competition = competition_map.get(h2_text)

        if competition is None:
            continue

        rows = table.find_all("tr")

        if not rows:
            continue

        # Identify actual match tables rather than standings/stat tables.
        header_cells = rows[0].find_all(["th", "td"])

        headers = [
            clean_text(
                cell.get_text(" ", strip=True)
            )
            for cell in header_cells
        ]

        normalized_headers = [
            header.lower()
            .replace("–", "-")
            .replace("—", "-")
            for header in headers
        ]

        has_date = any(
            header == "date"
            for header in normalized_headers
        )

        has_opponent = any(
            "opponent" in header
            for header in normalized_headers
        )

        has_venue = any(
            header in {"h / a", "h/a"}
            for header in normalized_headers
        )

        has_result = any(
            "result" in header
            for header in normalized_headers
        )

        if not (
            has_date
            and has_opponent
            and has_venue
            and has_result
        ):
            continue

        # Locate columns dynamically.
        date_idx = next(
            i for i, header in enumerate(normalized_headers)
            if header == "date"
        )

        opponent_idx = next(
            i for i, header in enumerate(normalized_headers)
            if "opponent" in header
        )

        venue_idx = next(
            i for i, header in enumerate(normalized_headers)
            if header in {"h / a", "h/a"}
        )

        result_idx = next(
            i for i, header in enumerate(normalized_headers)
            if "result" in header
        )

        for row in rows[1:]:

            cells = row.find_all(["th", "td"])

            values = [
                clean_text(
                    cell.get_text(" ", strip=True)
                )
                for cell in cells
            ]

            required_max = max(
                date_idx,
                opponent_idx,
                venue_idx,
                result_idx,
            )

            if len(values) <= required_max:
                continue

            date_text = values[date_idx]
            opponent = values[opponent_idx]
            venue = values[venue_idx]
            raw_result = values[result_idx]

            # Parse dates such as "14 August 2021".
            date = pd.to_datetime(
                date_text,
                dayfirst=True,
                errors="coerce",
            )

            if pd.isna(date):
                continue

            # Ignore scheduled/unplayed rows.
            if not re.search(
                r"\d+\s*[–—-]\s*\d+",
                raw_result,
            ):
                continue

            fixtures.append({
                "season": season,
                "date": date,
                "team": "Manchester United",
                "opponent": opponent,
                "competition": competition,
                "venue": venue,
                "raw_result": raw_result,
                "source": url,
            })

    return pd.DataFrame(fixtures)


# ------------------------------------------------------------
# Collect both relevant season pages
# ------------------------------------------------------------

parts = []

for season, url in man_utd_second_seasons.items():

    season_df = collect_man_utd_second_season(
        season,
        url,
    )

    print(
        f"{season}: "
        f"{len(season_df)} played official fixtures "
        f"on full season page"
    )

    parts.append(season_df)


nonempty_parts = [
    part
    for part in parts
    if not part.empty
]

if not nonempty_parts:
    raise ValueError(
        "No Manchester United official fixtures were collected."
    )


man_utd_second_fixtures = (
    pd.concat(
        nonempty_parts,
        ignore_index=True,
    )
    .sort_values(
        ["date", "competition", "opponent"]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Restrict to Ronaldo's second Manchester United tenure
# ------------------------------------------------------------

man_utd_second_fixtures = (
    man_utd_second_fixtures[
        man_utd_second_fixtures["date"].between(
            pd.Timestamp("2021-08-27"),
            pd.Timestamp("2022-11-22"),
        )
    ]
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Audit
# ------------------------------------------------------------

duplicate_mask = (
    man_utd_second_fixtures.duplicated(
        subset=[
            "date",
            "opponent",
            "competition",
        ],
        keep=False,
    )
)


print(
    "\nMANCHESTER UNITED SECOND SPELL — COMPETITIVE FIXTURE COLLECTION"
)
print("=" * 100)

print(
    f"Total official fixtures inside tenure: "
    f"{len(man_utd_second_fixtures)}"
)

print(
    f"Date range: "
    f"{man_utd_second_fixtures['date'].min().date()} "
    f"→ "
    f"{man_utd_second_fixtures['date'].max().date()}"
)

print(
    f"Duplicate candidate rows: "
    f"{duplicate_mask.sum()}"
)

print(
    f"Missing dates: "
    f"{man_utd_second_fixtures['date'].isna().sum()}"
)

print(
    f"Missing opponents: "
    f"{man_utd_second_fixtures['opponent'].isna().sum()}"
)

print(
    f"Missing results: "
    f"{man_utd_second_fixtures['raw_result'].isna().sum()}"
)


print("\nFixtures by season:")
print(
    man_utd_second_fixtures[
        "season"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)


print("\nFixtures by competition:")
print(
    man_utd_second_fixtures[
        "competition"
    ]
    .value_counts()
    .to_string()
)


print("\nSeason × competition:")
print(
    pd.crosstab(
        man_utd_second_fixtures["season"],
        man_utd_second_fixtures["competition"],
    ).to_string()
)


print("\nFirst 5 fixtures inside tenure:")
print(
    man_utd_second_fixtures[
        [
            "date",
            "opponent",
            "competition",
            "venue",
            "raw_result",
        ]
    ]
    .head()
    .to_string(index=False)
)


print("\nLast 10 fixtures inside tenure:")
print(
    man_utd_second_fixtures[
        [
            "date",
            "opponent",
            "competition",
            "venue",
            "raw_result",
        ]
    ]
    .tail(10)
    .to_string(index=False)
)
```

```python
# Extract Al-Nassr 2023-24 from actual footballbox row structure

def extract_al_nassr_footballbox(box, competition):

    rows = box.find_all("tr")

    if not rows:
        return None

    # First row contains:
    # [date (+ possible round), home team, score, away team, location]
    first_cells = rows[0].find_all(
        ["th", "td"],
        recursive=False,
    )

    values = [
        clean_text(
            cell.get_text(" ", strip=True)
        )
        for cell in first_cells
    ]

    if len(values) < 4:
        return None

    date_round_text = values[0]
    home_team = values[1]
    raw_result = values[2]
    away_team = values[3]

    # Extract only the calendar date.
    # Example: "14 August 2023 1" -> "14 August 2023"
    date_match = re.search(
        r"\b(\d{1,2}\s+[A-Za-z]+\s+\d{4})\b",
        date_round_text,
    )

    if date_match is None:
        return None

    date = pd.to_datetime(
        date_match.group(1),
        dayfirst=True,
        errors="coerce",
    )

    if pd.isna(date):
        return None

    # Only played matches
    score_match = re.search(
        r"(\d+)\s*[–—-]\s*(\d+)",
        raw_result,
    )

    if score_match is None:
        return None

    home_goals = int(score_match.group(1))
    away_goals = int(score_match.group(2))

    # Robust team-name normalization
    home_key = re.sub(
        r"\s+",
        " ",
        normalize_match_text(home_team),
    ).strip()

    away_key = re.sub(
        r"\s+",
        " ",
        normalize_match_text(away_team),
    ).strip()

    al_nassr_keys = {
        "al nassr",
        "al nassr fc",
    }

    if home_key in al_nassr_keys:

        opponent = away_team
        venue = "H"

        team_goals = home_goals
        opponent_goals = away_goals

    elif away_key in al_nassr_keys:

        opponent = home_team
        venue = "A"

        team_goals = away_goals
        opponent_goals = home_goals

    else:
        return None

    return {
        "season": "2023-24",
        "date": date,
        "team": "Al-Nassr",
        "opponent": opponent,
        "competition": competition,
        "venue": venue,
        "team_goals": team_goals,
        "opponent_goals": opponent_goals,
        "raw_result": raw_result,
        "source": al_nassr_seasons["2023-24"],
    }


# ------------------------------------------------------------
# Extract official competitions
# ------------------------------------------------------------

aln_2324_rows = []

for box in boxes:

    h2 = box.find_previous("h2")
    h3 = box.find_previous("h3")

    if h2 is None or h3 is None:
        continue

    h2_text = clean_text(
        h2.get_text(" ", strip=True)
    ).lower()

    h3_text = clean_text(
        h3.get_text(" ", strip=True)
    ).lower()

    if h2_text != "competitions":
        continue

    competition = competition_map.get(
        h3_text
    )

    if competition is None:
        continue

    row = extract_al_nassr_footballbox(
        box,
        competition,
    )

    if row is not None:
        aln_2324_rows.append(row)


# Fail clearly instead of producing another KeyError
if not aln_2324_rows:
    raise ValueError(
        "No official Al-Nassr 2023-24 fixtures were extracted."
    )


aln_2324 = (
    pd.DataFrame(aln_2324_rows)
    .sort_values(
        ["date", "competition", "opponent"]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Audit
# ------------------------------------------------------------

duplicate_mask = aln_2324.duplicated(
    subset=[
        "date",
        "opponent",
        "competition",
    ],
    keep=False,
)


print(
    "AL-NASSR 2023-24 — OFFICIAL FIXTURE COLLECTION"
)
print("=" * 110)

print(
    f"Official competitive fixtures: "
    f"{len(aln_2324)}"
)

print(
    f"Date range: "
    f"{aln_2324['date'].min().date()} "
    f"→ {aln_2324['date'].max().date()}"
)

print(
    f"Duplicate candidate rows: "
    f"{duplicate_mask.sum()}"
)

print(
    f"Missing opponents: "
    f"{aln_2324['opponent'].isna().sum()}"
)

print(
    f"Missing results: "
    f"{aln_2324['raw_result'].isna().sum()}"
)


print("\nFixtures by competition:")
print(
    aln_2324["competition"]
    .value_counts()
    .to_string()
)


print("\nFirst 10 official fixtures:")
print(
    aln_2324[
        [
            "date",
            "opponent",
            "competition",
            "venue",
            "team_goals",
            "opponent_goals",
        ]
    ]
    .head(10)
    .to_string(index=False)
)


print("\nLast 10 official fixtures:")
print(
    aln_2324[
        [
            "date",
            "opponent",
            "competition",
            "venue",
            "team_goals",
            "opponent_goals",
        ]
    ]
    .tail(10)
    .to_string(index=False)
)
```

```python
#mCollect Al-Nassr 2024-25 official fixtures from Transfermarkt

tm_competition_map = {
    "Saudi Pro League": "Pro League",
    "AFC Champions League Elite": "AFC Champions League Elite",
    "King's Cup": "King Cup",
    "Saudi Super Cup": "Saudi Super Cup",
}


def collect_al_nassr_transfermarkt_season(
    season_label,
    season_id,
):

    url = (
        "https://www.transfermarkt.com/al-nassr-fc/"
        f"spielplan/verein/18544/saison_id/{season_id}"
    )

    response = requests.get(
        url,
        headers={
            "User-Agent": (
                "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                "AppleWebKit/537.36 (KHTML, like Gecko) "
                "Chrome/120.0.0.0 Safari/537.36"
            )
        },
        timeout=30,
    )

    response.raise_for_status()

    soup = BeautifulSoup(
        response.text,
        "html.parser",
    )

    fixtures = []

    # Competition name appears in the H2 immediately before its table.
    for table in soup.find_all("table"):

        heading = table.find_previous("h2")

        if heading is None:
            continue

        heading_text = clean_text(
            heading.get_text(
                " ",
                strip=True,
            )
        )

        competition = tm_competition_map.get(
            heading_text
        )

        if competition is None:
            continue

        rows = table.find_all("tr")

        if len(rows) < 2:
            continue

        # Confirm this is a match table.
        header_values = [
            clean_text(
                cell.get_text(
                    " ",
                    strip=True,
                )
            )
            for cell in rows[0].find_all(
                ["th", "td"]
            )
        ]

        if not (
            "Date" in header_values
            and "Venue" in header_values
            and "Opponent" in header_values
            and "Result" in header_values
        ):
            continue

        for row in rows[1:]:

            cells = row.find_all("td")

            values = [
                clean_text(
                    cell.get_text(
                        " ",
                        strip=True,
                    )
                )
                for cell in cells
            ]

            # Observed Transfermarkt structure:
            #
            # Matchday, Date, Time, Venue, Ranking,
            # blank/logo, Opponent, System, Attendance, Result

            if len(values) < 10:
                continue

            date_text = values[1]
            venue = values[3]
            opponent = values[6]
            raw_result = values[-1]

            date = pd.to_datetime(
                date_text,
                format="%a %d/%m/%Y",
                errors="coerce",
            )

            if pd.isna(date):
                continue

            # Only played fixtures
            score_match = re.search(
                r"(\d+)\s*[:–—-]\s*(\d+)",
                raw_result,
            )

            if score_match is None:
                continue

            first_goals = int(
                score_match.group(1)
            )

            second_goals = int(
                score_match.group(2)
            )

            # Transfermarkt displays score in home-team order.
            if venue == "H":

                team_goals = first_goals
                opponent_goals = second_goals

            elif venue == "A":

                team_goals = second_goals
                opponent_goals = first_goals

            else:
                # Preserve unexpected venue values for inspection.
                team_goals = pd.NA
                opponent_goals = pd.NA

            fixtures.append({
                "season": season_label,
                "date": date,
                "team": "Al-Nassr",
                "opponent": opponent,
                "competition": competition,
                "venue": venue,
                "team_goals": team_goals,
                "opponent_goals": opponent_goals,
                "raw_result": raw_result,
                "source": url,
            })

    return (
        pd.DataFrame(fixtures)
        .sort_values(
            ["date", "competition", "opponent"]
        )
        .reset_index(drop=True)
    )


# ------------------------------------------------------------
# Test on 2024-25
# ------------------------------------------------------------

aln_2425 = collect_al_nassr_transfermarkt_season(
    season_label="2024-25",
    season_id=2024,
)


duplicate_mask = aln_2425.duplicated(
    subset=[
        "date",
        "opponent",
        "competition",
    ],
    keep=False,
)


print(
    "AL-NASSR 2024-25 — TRANSFERMARKT FIXTURE COLLECTION"
)
print("=" * 110)

print(
    f"Official competitive fixtures: "
    f"{len(aln_2425)}"
)

print(
    f"Date range: "
    f"{aln_2425['date'].min().date()} "
    f"→ {aln_2425['date'].max().date()}"
)

print(
    f"Duplicate candidate rows: "
    f"{duplicate_mask.sum()}"
)

print(
    f"Missing opponents: "
    f"{aln_2425['opponent'].isna().sum()}"
)

print(
    f"Unexpected venue values: "
    f"{(~aln_2425['venue'].isin(['H', 'A'])).sum()}"
)


print("\nFixtures by competition:")
print(
    aln_2425[
        "competition"
    ]
    .value_counts()
    .to_string()
)


print("\nFirst 10 fixtures:")
print(
    aln_2425[
        [
            "date",
            "opponent",
            "competition",
            "venue",
            "team_goals",
            "opponent_goals",
        ]
    ]
    .head(10)
    .to_string(index=False)
)


print("\nLast 10 fixtures:")
print(
    aln_2425[
        [
            "date",
            "opponent",
            "competition",
            "venue",
            "team_goals",
            "opponent_goals",
        ]
    ]
    .tail(10)
    .to_string(index=False)
)
```

```python
# Audit ALL Transfermarkt competition tables before trusting fixture totals

tm_seasons = {
    "2022-23": 2022,
    "2023-24": 2023,
    "2024-25": 2024,
    "2025-26": 2025,
    "2026-27": 2026,
}


def audit_tm_season(season_label, season_id):

    url = (
        "https://www.transfermarkt.com/al-nassr-fc/"
        f"spielplan/verein/18544/saison_id/{season_id}"
    )

    response = requests.get(
        url,
        headers={
            "User-Agent": (
                "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                "AppleWebKit/537.36 (KHTML, like Gecko) "
                "Chrome/120.0.0.0 Safari/537.36"
            )
        },
        timeout=30,
    )

    response.raise_for_status()

    soup = BeautifulSoup(
        response.text,
        "html.parser",
    )

    print(f"\n{season_label}")
    print("=" * 110)
    print("URL:", url)

    match_table_number = 0

    for table_number, table in enumerate(
        soup.find_all("table"),
        start=1,
    ):

        rows = table.find_all("tr")

        if not rows:
            continue

        header = [
            clean_text(
                cell.get_text(" ", strip=True)
            )
            for cell in rows[0].find_all(
                ["th", "td"]
            )
        ]

        # Only fixture-style tables
        if not (
            "Date" in header
            and "Opponent" in header
            and "Result" in header
        ):
            continue

        match_table_number += 1

        # Find closest heading above table
        heading = table.find_previous(
            ["h1", "h2", "h3"]
        )

        heading_text = (
            clean_text(
                heading.get_text(
                    " ",
                    strip=True,
                )
            )
            if heading
            else "UNKNOWN"
        )

        played = 0
        unplayed = 0

        dates = []
        sample_rows = []

        for row in rows[1:]:

            cells = row.find_all("td")

            values = [
                clean_text(
                    cell.get_text(
                        " ",
                        strip=True,
                    )
                )
                for cell in cells
            ]

            if not values:
                continue

            # Result is last cell in the observed fixture tables
            result_text = values[-1]

            score_match = re.search(
                r"\d+\s*[:–—-]\s*\d+",
                result_text,
            )

            if score_match:
                played += 1
            else:
                unplayed += 1

            # Date is position 1 in observed Transfermarkt structure
            if len(values) > 1:

                parsed_date = pd.to_datetime(
                    values[1],
                    format="%a %d/%m/%Y",
                    errors="coerce",
                )

                if not pd.isna(parsed_date):
                    dates.append(parsed_date)

            if len(sample_rows) < 2:
                sample_rows.append(values)

        print(
            f"\nMATCH TABLE {match_table_number}"
        )

        print(
            f"Heading: {heading_text}"
        )

        print(
            f"Rows: {len(rows) - 1} | "
            f"Played: {played} | "
            f"Unplayed/other: {unplayed}"
        )

        if dates:

            print(
                f"Date range: "
                f"{min(dates).date()} "
                f"→ {max(dates).date()}"
            )

        print("Samples:")

        for sample in sample_rows:
            print(" ", sample)


print(
    "AL-NASSR — COMPLETE TRANSFERMARKT TABLE AUDIT"
)
print("=" * 110)

for season_label, season_id in tm_seasons.items():

    audit_tm_season(
        season_label,
        season_id,
    )
```

```python
# Initial Al-Nassr reconciliation against canonical appearances

canonical_path = (
    "../data/processed/"
    "ronaldo_1000_ml_canonical_final.csv"
)

canonical = pd.read_csv(
    canonical_path,
    parse_dates=["date"],
)


# ------------------------------------------------------------
# 1. Isolate canonical Al-Nassr appearances
# ------------------------------------------------------------

canonical_aln = (
    canonical.loc[
        canonical["team"].eq("Al-Nassr")
        & canonical["date"].le("2026-08-28")
    ]
    .copy()
    .sort_values("date")
    .reset_index(drop=True)
)


print(
    "AL-NASSR — CANONICAL APPEARANCE RECONCILIATION"
)
print("=" * 110)

print(
    f"Canonical Al-Nassr appearances: "
    f"{len(canonical_aln)}"
)

print(
    f"Date range: "
    f"{canonical_aln['date'].min().date()} "
    f"→ {canonical_aln['date'].max().date()}"
)


print("\nCanonical competitions:")
print(
    canonical_aln["competition"]
    .value_counts()
    .to_string()
)


# ------------------------------------------------------------
# 2. Build conservative normalized matching keys
# ------------------------------------------------------------

def aln_match_key(value):

    value = normalize_match_text(
        str(value)
    )

    value = re.sub(
        r"\s+",
        " ",
        value,
    ).strip()

    return value


fixtures = aln_candidate.copy()
apps = canonical_aln.copy()


fixtures["opponent_key"] = (
    fixtures["opponent"]
    .map(aln_match_key)
)

apps["opponent_key"] = (
    apps["opponent"]
    .map(aln_match_key)
)


# Competition normalization ONLY for clearly equivalent labels.
competition_key_map = {
    "Pro League": "pro_league",
    "King Cup": "king_cup",
    "Saudi Super Cup": "saudi_super_cup",

    "AFC Champions League": "afc_champions_league",
    "AFC Champions League Elite": "afc_champions_league_elite",
    "AFC Champions League Two": "afc_champions_league_two",

    "Arab Club Champions Cup": "arab_club_champions_cup",
}


fixtures["competition_key"] = (
    fixtures["competition"]
    .map(competition_key_map)
    .fillna(
        fixtures["competition"]
        .map(aln_match_key)
    )
)

apps["competition_key"] = (
    apps["competition"]
    .map(aln_match_key)
)


# Show canonical labels before attempting to force equivalence.
print("\nCanonical competition keys:")
print(
    apps[
        [
            "competition",
            "competition_key",
        ]
    ]
    .drop_duplicates()
    .sort_values("competition")
    .to_string(index=False)
)


# ------------------------------------------------------------
# 3. First-pass exact matching:
#    date + normalized opponent
#
# Competition is deliberately NOT required yet because the
# canonical competition labels may differ from our fixture labels.
# ------------------------------------------------------------

fixture_lookup = {}

for idx, row in fixtures.iterrows():

    key = (
        row["date"].normalize(),
        row["opponent_key"],
    )

    fixture_lookup.setdefault(
        key,
        []
    ).append(idx)


matched_fixture_indices = set()
match_records = []
unmatched_apps = []


for app_idx, row in apps.iterrows():

    key = (
        row["date"].normalize(),
        row["opponent_key"],
    )

    candidates = [
        idx
        for idx in fixture_lookup.get(key, [])
        if idx not in matched_fixture_indices
    ]

    if len(candidates) == 1:

        fixture_idx = candidates[0]

        matched_fixture_indices.add(
            fixture_idx
        )

        match_records.append({
            "canonical_index": app_idx,
            "fixture_index": fixture_idx,
            "match_id": row["match_id"],
            "date": row["date"],
            "opponent": row["opponent"],
            "canonical_competition": row["competition"],
            "fixture_competition": fixtures.loc[
                fixture_idx,
                "competition",
            ],
            "status": "appeared_exact",
        })

    else:

        unmatched_apps.append({
            "canonical_index": app_idx,
            "match_id": row["match_id"],
            "date": row["date"],
            "opponent": row["opponent"],
            "competition": row["competition"],
            "candidate_count": len(candidates),
        })


matched = pd.DataFrame(
    match_records
)

unmatched = pd.DataFrame(
    unmatched_apps
)


# ------------------------------------------------------------
# 4. Audit
# ------------------------------------------------------------

print("\nINITIAL RECONCILIATION")
print("-" * 110)

print(
    f"Candidate team fixtures: "
    f"{len(fixtures)}"
)

print(
    f"Canonical appearances: "
    f"{len(apps)}"
)

print(
    f"Exact date+opponent matches: "
    f"{len(matched)}"
)

print(
    f"Unmatched canonical appearances: "
    f"{len(unmatched)}"
)

print(
    f"Candidate fixtures not yet linked: "
    f"{len(fixtures) - len(matched_fixture_indices)}"
)


if not unmatched.empty:

    print(
        "\nUNMATCHED CANONICAL APPEARANCES"
    )
    print("-" * 110)

    print(
        unmatched[
            [
                "date",
                "opponent",
                "competition",
                "candidate_count",
            ]
        ]
        .to_string(index=False)
    )


# ------------------------------------------------------------
# 5. Inspect competition-label disagreements among exact matches
# ------------------------------------------------------------

if not matched.empty:

    comp_pairs = (
        matched.groupby(
            [
                "canonical_competition",
                "fixture_competition",
            ]
        )
        .size()
        .reset_index(name="matches")
        .sort_values(
            "matches",
            ascending=False,
        )
    )

    print(
        "\nCANONICAL ↔ FIXTURE COMPETITION LABELS"
    )
    print("-" * 110)

    print(
        comp_pairs.to_string(
            index=False
        )
    )


# ------------------------------------------------------------
# 6. Inspect unmatched appearances against nearby fixtures
# ------------------------------------------------------------

if not unmatched.empty:

    print(
        "\nNEARBY FIXTURES FOR EACH UNMATCHED APPEARANCE"
    )
    print("=" * 110)

    for _, row in unmatched.iterrows():

        date = row["date"]

        nearby = fixtures.loc[
            fixtures["date"].between(
                date - pd.Timedelta(days=2),
                date + pd.Timedelta(days=2),
            ),
            [
                "date",
                "opponent",
                "competition",
                "venue",
                "team_goals",
                "opponent_goals",
            ],
        ]

        print(
            f"\nCanonical: "
            f"{date.date()} | "
            f"{row['opponent']} | "
            f"{row['competition']}"
        )

        if nearby.empty:

            print(
                "  No candidate fixture within ±2 days."
            )

        else:

            print(
                nearby.to_string(
                    index=False
                )
            )
```

```python
# Normalize Al-Nassr opponent aliases and rerun exact matching

aln_opponent_aliases = {
    # Saudi clubs
    "al ittihad fc": "al ittihad",
    "al wehda club": "al wehda",
    "al taawon": "al taawoun",
    "dhamk": "damac fc",
    "damac": "damac fc",
    "al baten": "al batin",
    "al khaleej saihat": "al khaleej",
    "al taee": "al tai",
    "al hazm": "al hazem",
    "al ahli jeddah": "al ahli",
    "al orubah": "al orobah",
    "al qadisiyah fc": "al qadsiah",

    # Arab Club Champions Cup
    "us monastirienne": "union monastirienne",
    "zamalek sc": "zamalek",
    "raja casablanca": "raja ca",

    # AFC competitions
    "shabab al ahli dubai": "shabab al ahli",
    "persepolis fc": "persepolis",
    "istiqlol": "istiklol",
    "al duhail sc": "al duhail",
    "al rayyan sc": "al rayyan",
    "al wasl fc": "al wasl",
    "yokohama f marinos": "yokohama f m",
    "kawasaki frontale": "kawasaki front",
    "al zawra a": "al zawraa sc",
    "al ahli doha": "al ahli sc",

    # Other
    "neom": "neom sc",
}


def aln_reconciliation_key(value):

    key = aln_match_key(value)

    return aln_opponent_aliases.get(
        key,
        key,
    )


fixtures = aln_candidate.copy()
apps = canonical_aln.copy()

fixtures["opponent_key"] = (
    fixtures["opponent"]
    .map(aln_reconciliation_key)
)

apps["opponent_key"] = (
    apps["opponent"]
    .map(aln_reconciliation_key)
)


# ------------------------------------------------------------
# Exact date + normalized opponent reconciliation
# ------------------------------------------------------------

fixture_lookup = {}

for fixture_idx, row in fixtures.iterrows():

    key = (
        row["date"].normalize(),
        row["opponent_key"],
    )

    fixture_lookup.setdefault(
        key,
        []
    ).append(fixture_idx)


matched_fixture_indices = set()
match_records = []
unmatched_records = []


for app_idx, row in apps.iterrows():

    key = (
        row["date"].normalize(),
        row["opponent_key"],
    )

    candidates = [
        idx
        for idx in fixture_lookup.get(key, [])
        if idx not in matched_fixture_indices
    ]

    if len(candidates) == 1:

        fixture_idx = candidates[0]

        matched_fixture_indices.add(
            fixture_idx
        )

        match_records.append({
            "canonical_index": app_idx,
            "fixture_index": fixture_idx,
            "match_id": row["match_id"],
            "date": row["date"],
            "canonical_opponent": row["opponent"],
            "fixture_opponent": fixtures.loc[
                fixture_idx,
                "opponent",
            ],
            "canonical_competition": row["competition"],
            "fixture_competition": fixtures.loc[
                fixture_idx,
                "competition",
            ],
            "status": "appeared_exact",
        })

    else:

        unmatched_records.append({
            "canonical_index": app_idx,
            "match_id": row["match_id"],
            "date": row["date"],
            "opponent": row["opponent"],
            "competition": row["competition"],
            "candidate_count": len(candidates),
        })


matched = pd.DataFrame(
    match_records
)

unmatched = pd.DataFrame(
    unmatched_records
)


# ------------------------------------------------------------
# Audit
# ------------------------------------------------------------

print(
    "AL-NASSR — ALIAS-NORMALIZED RECONCILIATION"
)
print("=" * 110)

print(
    f"Candidate fixtures: {len(fixtures)}"
)

print(
    f"Canonical appearances: {len(apps)}"
)

print(
    f"Matched appearances: {len(matched)}"
)

print(
    f"Unmatched canonical appearances: {len(unmatched)}"
)

print(
    f"Candidate fixtures not linked to an appearance: "
    f"{len(fixtures) - len(matched_fixture_indices)}"
)

print(
    f"Unique canonical match IDs linked: "
    f"{matched['match_id'].nunique() if not matched.empty else 0}"
)


if not unmatched.empty:

    print(
        "\nSTILL UNMATCHED"
    )
    print("-" * 110)

    print(
        unmatched[
            [
                "date",
                "opponent",
                "competition",
                "candidate_count",
            ]
        ]
        .to_string(index=False)
    )


# ------------------------------------------------------------
# Show all alias transformations that actually participated
# ------------------------------------------------------------

alias_matches = matched.loc[
    matched["canonical_opponent"]
    != matched["fixture_opponent"]
].copy()

print(
    "\nMATCHED OPPONENT-NAME DIFFERENCES"
)
print("-" * 110)

print(
    alias_matches[
        [
            "canonical_opponent",
            "fixture_opponent",
        ]
    ]
    .drop_duplicates()
    .sort_values(
        "canonical_opponent"
    )
    .to_string(index=False)
)


# ------------------------------------------------------------
# Candidate nonappearances by season
# ------------------------------------------------------------

nonappearance_candidates = fixtures.loc[
    ~fixtures.index.isin(
        matched_fixture_indices
    )
].copy()


print(
    "\nCANDIDATE NONAPPEARANCE FIXTURES BY SEASON"
)
print("-" * 110)

print(
    nonappearance_candidates[
        "season"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)
```

```python
# Apply Portugal tenure boundary and validate played results

PORTUGAL_START = pd.Timestamp("2003-08-20")
HISTORICAL_CUTOFF = pd.Timestamp("2026-08-28")


def extract_played_score(values):
    """
    Extract the final-score field from a Transfermarkt row.

    Reject:
    - kickoff times such as 8:45 PM
    - future placeholders such as -:-
    """

    for value in reversed(values):

        value = value.strip()

        # A played football score must consist only of
        # two integer scores separated by :, -, –, or —.
        if re.fullmatch(
            r"\d{1,2}\s*[:–—-]\s*\d{1,2}",
            value,
        ):
            return value

    return None


portugal_raw["played_result"] = (
    portugal_raw["row_values"]
    .apply(extract_played_score)
)


# Exact Ronaldo/forecast historical window
portugal_tenure = (
    portugal_raw.loc[
        portugal_raw["date"].between(
            PORTUGAL_START,
            HISTORICAL_CUTOFF,
            inclusive="both",
        )
    ]
    .copy()
    .sort_values("date")
    .reset_index(drop=True)
)


played_portugal = (
    portugal_tenure.loc[
        portugal_tenure["played_result"].notna()
    ]
    .copy()
    .reset_index(drop=True)
)


unplayed_in_window = (
    portugal_tenure.loc[
        portugal_tenure["played_result"].isna()
    ]
    .copy()
)


before_debut = portugal_raw.loc[
    portugal_raw["date"] < PORTUGAL_START
].copy()


after_cutoff = portugal_raw.loc[
    portugal_raw["date"] > HISTORICAL_CUTOFF
].copy()


print(
    "PORTUGAL — HISTORICAL TENURE FIXTURE AUDIT"
)
print("=" * 100)

print(
    f"Raw Transfermarkt rows: "
    f"{len(portugal_raw)}"
)

print(
    f"Before Ronaldo debut: "
    f"{len(before_debut)}"
)

print(
    f"After historical cutoff: "
    f"{len(after_cutoff)}"
)

print(
    f"Rows inside tenure window: "
    f"{len(portugal_tenure)}"
)

print(
    f"Played fixtures inside window: "
    f"{len(played_portugal)}"
)

print(
    f"Unplayed fixtures inside window: "
    f"{len(unplayed_in_window)}"
)


print(
    "\nPlayed date range:",
    played_portugal["date"].min().date(),
    "→",
    played_portugal["date"].max().date(),
)


print("\nPlayed fixtures by year:")
print(
    played_portugal["date"]
    .dt.year
    .value_counts()
    .sort_index()
    .to_string()
)


print("\nPlayed fixtures by competition:")
print(
    played_portugal["competition_raw"]
    .value_counts()
    .to_string()
)


print("\nBefore-debut rows:")
for _, row in before_debut.iterrows():
    print(
        row["date"].date(),
        "|",
        row["competition_raw"],
        "|",
        row["played_result"],
    )


print("\nAfter-cutoff rows:")
for _, row in after_cutoff.iterrows():
    print(
        row["date"].date(),
        "|",
        row["competition_raw"],
        "|",
        row["played_result"],
        "|",
        row["row_values"],
    )


if len(unplayed_in_window):
    print("\nWARNING — unplayed rows inside historical window:")

    for _, row in unplayed_in_window.iterrows():
        print(
            row["date"].date(),
            "|",
            row["competition_raw"],
            "|",
            row["row_values"],
        )
else:
    print(
        "\n✓ Every fixture inside the historical "
        "tenure window has a played final score."
    )
```

```python
# Correctly recognize ordinary, AET and penalty-shootout results

def extract_completed_result(values):

    for value in reversed(values):

        value = value.strip()

        # Ordinary final score
        if re.fullmatch(
            r"\d{1,2}\s*[:–—-]\s*\d{1,2}",
            value,
        ):
            return value

        # Completed after extra time
        if re.fullmatch(
            r"\d{1,2}\s*[:–—-]\s*\d{1,2}\s+AET",
            value,
            flags=re.IGNORECASE,
        ):
            return value

        # Completed via penalty shootout
        if re.fullmatch(
            r"\d{1,2}\s*[:–—-]\s*\d{1,2}\s+on\s+pens",
            value,
            flags=re.IGNORECASE,
        ):
            return value

    return None


portugal_tenure["completed_result"] = (
    portugal_tenure["row_values"]
    .apply(extract_completed_result)
)


played_portugal = (
    portugal_tenure.loc[
        portugal_tenure[
            "completed_result"
        ].notna()
    ]
    .copy()
    .sort_values("date")
    .reset_index(drop=True)
)


still_unresolved = (
    portugal_tenure.loc[
        portugal_tenure[
            "completed_result"
        ].isna()
    ]
    .copy()
)


print(
    "PORTUGAL — COMPLETED HISTORICAL FIXTURE UNIVERSE"
)
print("=" * 100)

print(
    f"Tenure-window rows: "
    f"{len(portugal_tenure)}"
)

print(
    f"Completed fixtures: "
    f"{len(played_portugal)}"
)

print(
    f"Still unresolved: "
    f"{len(still_unresolved)}"
)

print(
    f"Date range: "
    f"{played_portugal['date'].min().date()} "
    f"→ "
    f"{played_portugal['date'].max().date()}"
)


print("\nCompleted fixtures by year:")
print(
    played_portugal["date"]
    .dt.year
    .value_counts()
    .sort_index()
    .to_string()
)


print("\nCompleted fixtures by competition:")
print(
    played_portugal[
        "competition_raw"
    ]
    .value_counts()
    .to_string()
)


print("\nSpecial-result fixtures:")
special = played_portugal.loc[
    played_portugal[
        "completed_result"
    ].str.contains(
        r"AET|pens",
        case=False,
        regex=True,
        na=False,
    )
]

for _, row in special.iterrows():

    print(
        row["date"].date(),
        "|",
        row["competition_raw"],
        "|",
        row["completed_result"],
        "|",
        row["row_values"],
    )


if len(still_unresolved):

    print("\nUNRESOLVED:")

    for _, row in still_unresolved.iterrows():

        print(
            row["date"].date(),
            "|",
            row["competition_raw"],
            "|",
            row["row_values"],
        )

else:

    print(
        "\n✓ Every Portugal fixture from "
        "Ronaldo's debut through 2026-08-28 "
        "is recognized as completed."
    )


assert len(played_portugal) == 296
assert len(still_unresolved) == 0
```

```python
# Parse Portugal fixture fields from Transfermarkt rows

def parse_portugal_fixture(row):

    values = row["row_values"]

    # Locate the date inside the raw row.
    date_idx = None

    for i, value in enumerate(values):

        cleaned = re.sub(
            r"^[A-Za-z]{2,4}\s+",
            "",
            value.strip(),
        )

        parsed = pd.to_datetime(
            cleaned,
            dayfirst=True,
            errors="coerce",
        )

        if (
            not pd.isna(parsed)
            and re.search(
                r"\d{1,2}/\d{1,2}/\d{4}",
                value,
            )
        ):
            date_idx = i
            break

    if date_idx is None:
        raise ValueError(
            f"Could not locate date: {values}"
        )

    # Transfermarkt structure after date:
    # date, time, H/A/N, blank, blank, opponent, ...
    venue_idx = date_idx + 2
    opponent_idx = date_idx + 5

    venue = values[venue_idx].strip()
    opponent = values[opponent_idx].strip()

    result_raw = row["completed_result"].strip()

    if re.search(
        r"on\s+pens",
        result_raw,
        flags=re.IGNORECASE,
    ):
        result_type = "pens"

    elif re.search(
        r"AET",
        result_raw,
        flags=re.IGNORECASE,
    ):
        result_type = "aet"

    else:
        result_type = "normal"

    return pd.Series({
        "date": row["date"],
        "team": "Portugal",
        "opponent": opponent,
        "competition_raw": row["competition_raw"],
        "venue": venue,
        "result_raw": result_raw,
        "result_type": result_type,
        "source": "Transfermarkt",
    })


portugal_fixtures = (
    played_portugal
    .apply(
        parse_portugal_fixture,
        axis=1,
    )
    .sort_values("date")
    .reset_index(drop=True)
)


print(
    "PORTUGAL — PARSED FIXTURE UNIVERSE"
)
print("=" * 100)

print(
    "Shape:",
    portugal_fixtures.shape,
)

print(
    "Date range:",
    portugal_fixtures["date"].min().date(),
    "→",
    portugal_fixtures["date"].max().date(),
)

print(
    "Unique dates:",
    portugal_fixtures["date"].nunique(),
)

print(
    "\nVenue counts:"
)
print(
    portugal_fixtures["venue"]
    .value_counts(dropna=False)
    .to_string()
)

print(
    "\nResult types:"
)
print(
    portugal_fixtures["result_type"]
    .value_counts()
    .to_string()
)

print(
    "\nMissing opponents:",
    portugal_fixtures["opponent"]
    .eq("")
    .sum(),
)

print(
    "Unique opponents:",
    portugal_fixtures["opponent"]
    .nunique(),
)


print("\nFirst 10 fixtures:")
print(
    portugal_fixtures.head(10)
    .to_string(index=False)
)


print("\nLast 10 fixtures:")
print(
    portugal_fixtures.tail(10)
    .to_string(index=False)
)


print("\nSpecial-result fixtures:")
print(
    portugal_fixtures.loc[
        portugal_fixtures[
            "result_type"
        ].ne("normal")
    ]
    .to_string(index=False)
)


# Integrity checks
assert len(portugal_fixtures) == 296
assert portugal_fixtures["date"].nunique() == 296
assert portugal_fixtures["opponent"].ne("").all()

print(
    "\n✓ 296 Portugal senior fixtures "
    "parsed successfully."
)
```

## Standardize the seven tenure datasets in memory

```python
# Standardize all seven tenure datasets in memory
# No files are written in this step.

standardized_parts = []


for tenure, df_original in tenure_data.items():

    df = df_original.copy()

    # ---------------------------------------------
    # Canonical match ID
    # ---------------------------------------------
    if "canonical_match_id" in df.columns:
        canonical_id = df["canonical_match_id"]

    elif "match_id" in df.columns:
        canonical_id = df["match_id"]

    else:
        raise ValueError(
            f"{tenure}: no canonical match ID column found."
        )


    # ---------------------------------------------
    # Optional season
    # ---------------------------------------------
    if "season" in df.columns:
        season = df["season"]
    else:
        season = pd.Series(
            pd.NA,
            index=df.index,
            dtype="object",
        )


    # ---------------------------------------------
    # Build common schema
    # ---------------------------------------------
    standardized = pd.DataFrame({
        "date":
            pd.to_datetime(df["date"]),

        "team":
            df["team"],

        "opponent":
            df["opponent"],

        "competition":
            df["competition"],

        "venue":
            df["venue"],

        "appeared":
            df["appeared"].astype(int),

        "canonical_match_id":
            canonical_id,

        "season":
            season,

        "tenure":
            tenure,
    })


    standardized_parts.append(
        standardized
    )


appearance_master = (
    pd.concat(
        standardized_parts,
        ignore_index=True,
    )
    .sort_values(
        [
            "date",
            "team",
            "opponent",
        ]
    )
    .reset_index(drop=True)
)


print(
    "RONALDO APPEARANCE MODEL — "
    "STANDARDIZED MASTER TABLE"
)
print("=" * 110)

print(
    "Shape:",
    appearance_master.shape,
)

print(
    "Date range:",
    appearance_master["date"].min().date(),
    "→",
    appearance_master["date"].max().date(),
)

print(
    "Eligible opportunities:",
    len(appearance_master),
)

print(
    "Appearances:",
    int(appearance_master["appeared"].sum()),
)

print(
    "Nonappearances:",
    int(appearance_master["appeared"].eq(0).sum()),
)

print(
    "Appearance rate:",
    round(
        appearance_master["appeared"].mean(),
        4,
    ),
)


print("\nColumns:")
print(
    appearance_master.columns.tolist()
)


print("\nMissing values:")
print(
    appearance_master
    .isna()
    .sum()
    .to_string()
)


print("\nBy tenure:")
tenure_summary = (
    appearance_master
    .groupby(
        "tenure",
        sort=False,
    )
    .agg(
        opportunities=("appeared", "size"),
        appearances=("appeared", "sum"),
    )
)

tenure_summary["nonappearances"] = (
    tenure_summary["opportunities"]
    - tenure_summary["appearances"]
)

tenure_summary["appearance_rate"] = (
    tenure_summary["appearances"]
    / tenure_summary["opportunities"]
)

print(
    tenure_summary.to_string()
)


# ---------------------------------------------
# Canonical appearance integrity
# ---------------------------------------------

appearance_rows = (
    appearance_master.loc[
        appearance_master["appeared"].eq(1)
    ]
)

nonappearance_rows = (
    appearance_master.loc[
        appearance_master["appeared"].eq(0)
    ]
)


print("\nCanonical ID integrity:")
print(
    "Appearance rows:",
    len(appearance_rows),
)

print(
    "Appearance rows with canonical ID:",
    appearance_rows[
        "canonical_match_id"
    ].notna().sum(),
)

print(
    "Unique canonical IDs:",
    appearance_rows[
        "canonical_match_id"
    ].nunique(),
)

print(
    "Duplicate canonical IDs:",
    appearance_rows[
        "canonical_match_id"
    ]
    .duplicated()
    .sum(),
)

print(
    "Nonappearance rows with canonical ID:",
    nonappearance_rows[
        "canonical_match_id"
    ].notna().sum(),
)


# ---------------------------------------------
# Final assertions
# ---------------------------------------------

assert appearance_master.shape == (1584, 9)

assert appearance_master["appeared"].sum() == 1333

assert appearance_master["appeared"].eq(0).sum() == 251

assert len(appearance_rows) == 1333

assert (
    appearance_rows[
        "canonical_match_id"
    ].notna().all()
)

assert (
    appearance_rows[
        "canonical_match_id"
    ].nunique()
    == 1333
)

assert (
    appearance_rows[
        "canonical_match_id"
    ].duplicated()
    .sum()
    == 0
)

assert (
    nonappearance_rows[
        "canonical_match_id"
    ].isna()
    .all()
)

print(
    "\n✓ Seven tenure schemas standardized."
)

print(
    "✓ All 1,333 appearance rows map "
    "one-to-one to the canonical history."
)

print(
    "✓ All 251 nonappearance rows remain "
    "independent eligible fixture opportunities."
)

print(
    "✓ No source CSV was modified."
)
```
